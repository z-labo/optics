## 1. Coordinate System and Polarization

Let the angle between the optic axis of the BBO crystal and the crystal surface normal be $\theta_c$. The coordinate system is defined such that the crystal surface normal is parallel to the $z$ axis and the optic axis lies in the $x-z$ plane.

Let the external incidence angles of the two fundamental beams relative to the surface normal be $\gamma_1'$ and $\gamma_2'$, respectively. The polarizations of both beams are aligned along the $y$ direction.

In this geometry, the optic axis $\mathbf{c}$, $\mathbf{k}_1(\omega)$, and $\mathbf{k}_2(\omega)$ all lie in the $x-z$ plane. The electric fields of the fundamental beams satisfy

$$\mathbf{E}_1,\mathbf{E}_2\parallel y$$

and are therefore perpendicular to this principal plane. Hence, both fundamental beams propagate inside the crystal as ordinary waves.

BBO is a negative uniaxial crystal, and Type-I SHG allows the interaction

$$o(\omega)+o(\omega)\rightarrow e(2\omega).$$

## 2. Refraction Angles Inside the Crystal

Let the refraction angles inside the crystal be $\gamma_1$ and $\gamma_2$. From Snell's law,

$$\gamma_1=\sin^{-1}\left(\frac{\sin\gamma_1'}{n_o(\omega)}\right)$$

$$\gamma_2=\sin^{-1}\left(\frac{\sin\gamma_2'}{n_o(\omega)}\right)$$

## 3. Momentum Conservation in Non-Collinear SHG

For non-collinear SHG, the momentum-conservation, or phase-matching, condition is

$$\mathbf{k}_1+\mathbf{k}_2=\mathbf{k}_3$$

When the two fundamental beams have the same wavelength and refractive index and propagate symmetrically at angles $\gamma$ with respect to the $z$ axis, the $x$ components of the two fundamental wave vectors cancel. The generated SH wave therefore propagates along the $z$ direction, with wave-vector magnitude

$$k_3=2k_{1,2}\cos\gamma$$

For the fundamental waves,

$$k_{1,2}=\frac{2\pi}{\lambda_\omega}n_o(\omega)$$

whereas for the second-harmonic wave,

$$k_3=\frac{2\pi}{\lambda_\omega/2}n_{\mathrm{e,eff}}(2\omega,\theta)$$

Combining these expressions gives

$$\frac{2\pi}{\lambda_\omega/2}n_{e,\mathrm{eff}}(2\omega,\theta)=\frac{4\pi}{\lambda_\omega}n_o(\omega)\cos\gamma$$

and therefore

$$n_{e,\mathrm{eff}}(2\omega,\theta)=n_o(\omega)\cos\gamma$$

This is the phase-matching condition required for non-collinear Type-I SHG.

For the collinear case,

$$\gamma=0,\qquad n_{e,\mathrm{eff}}(2\omega,\theta)=n_o(\omega)$$

## 4. Refractive Indices of BBO

The Sellmeier equations of BBO are

$$n_o^2(\lambda)=2.7359+\frac{0.01878}{\lambda^2-0.01822}-0.01354\lambda^2$$

$$n_e^2(\lambda)=2.3753+\frac{0.01224}{\lambda^2-0.01667}-0.01516\lambda^2$$

where $\lambda$ is expressed in $\mu\mathrm{m}$. The resulting refractive indices are

| Quantity | Calculated value |
|---|---:|
| $n_o(808\,\mathrm{nm})$ | 1.66032 |
| $n_o(404\,\mathrm{nm})$ | 1.69210 |
| $n_e(404\,\mathrm{nm})$ | 1.56727 |

## 5. Effective Refractive Index of the Extraordinary SH Wave

Let $\theta$ be the angle between the extraordinary SH wave vector and the optic axis. The effective extraordinary refractive index is

$$\frac{1}{n_{e,\mathrm{eff}}^2(2\omega,\theta)}=\frac{\cos^2\theta}{n_o^2(2\omega)}+\frac{\sin^2\theta}{n_e^2(2\omega)}$$

If the SH beam propagates along the $z$ direction, then $\theta=\theta_c$. Therefore, the phase-matching condition requires

$$\frac{1}{\sqrt{\dfrac{\cos^2\theta}{n_o^2(2\omega)}+\dfrac{\sin^2\theta}{n_e^2(2\omega)}}}=n_o(\omega)\cos\gamma$$

## 6. Angles for the Present Geometry

The internal angle satisfying the above condition is

$$\gamma=1.57^\circ$$

and the corresponding incidence angle in air is

$$\gamma'=2.60^\circ$$

For a lens with focal length $f=200\,\mathrm{mm}$ and a beam offset of $12.5\,\mathrm{mm}$ from the optical axis,

$$f=200\,\mathrm{mm},\qquad \tan\gamma'=0.0625,\qquad \gamma'=3.58^\circ$$

Thus, strictly speaking, this angle is not the exact phase-matching angle. However, the difference is small and the crystal is thin, so the configuration can be regarded as being close to the phase-matching condition.

When a lens with focal length $75\,\mathrm{mm}$ is used, the offset between each input beam and the optical axis should be reduced to

$$x=f\tan(2.60^\circ)\simeq 3.4\,\mathrm{mm}\qquad (f=75\,\mathrm{mm})$$
