## 1. 座標系と偏光

BBO結晶のoptic axisとcrystal surface normalの間の角度を$\theta_c$とする。座標系はcrystal surface normalが$z$軸に平行となるように設定し、optic axisが$x-z$平面内に存在するようにする。

2本のfundamental beamの外部入射角を、surface normalに対してそれぞれ$\gamma_1'$、$\gamma_2'$とする。また、両方のbeamの偏光を$y$方向に揃える。

この場合、optic axis $\mathbf{c}$、$\mathbf{k}_1(\omega)$、$\mathbf{k}_2(\omega)$はすべて$x-z$平面内に存在する。Fundamental beamの電場は

$$\mathbf{E}_1,\mathbf{E}_2\parallel y$$

であるため、このprincipal planeに垂直である。したがって、2本のfundamental beamはいずれもordinary waveとして結晶内部を伝搬する。

BBOはnegative uniaxial crystalであり、Type-I SHGでは

$$o(\omega)+o(\omega)\rightarrow e(2\omega)$$

の相互作用が可能である。

## 2. 結晶内部での屈折角

結晶内部での屈折角をそれぞれ $\gamma_1$、$\gamma_2$ とする。Snellの法則より、

$$\gamma_1=\sin^{-1}\left(\frac{\sin\gamma_1'}{n_o(\omega)}\right)$$

$$\gamma_2=\sin^{-1}\left(\frac{\sin\gamma_2'}{n_o(\omega)}\right)$$

## 3. 非共線SHGの運動量保存

非共線SHGにおけるmomentum conservation、すなわちphase-matching conditionは

$$\mathbf{k}_1+\mathbf{k}_2=\mathbf{k}_3$$

である。

2本のfundamental beamが同一の波長と屈折率を持ち、$z$軸に対して対称に角度$\gamma$をなす場合、2本のfundamental wave vectorの$x$成分は互いに打ち消し合い、SH waveは$z$方向に伝搬する。このとき、生成されるsecond-harmonic wave vectorの大きさは

$$k_3=2k_{1,2}\cos\gamma$$

ここで、fundamental waveについては

$$k_{1,2}=\frac{2\pi}{\lambda_\omega}n_o(\omega)$$

であり、second-harmonic waveについては

$$k_3=\frac{2\pi}{\lambda_\omega/2}n_{\mathrm{e,eff}}(2\omega,\theta)$$

である。以上の式を組み合わせると、

$$\frac{2\pi}{\lambda_\omega/2}n_{e,\mathrm{eff}}(2\omega,\theta)=\frac{4\pi}{\lambda_\omega}n_o(\omega)\cos\gamma$$

すなわち、

$$n_{e,\mathrm{eff}}(2\omega,\theta)=n_o(\omega)\cos\gamma$$

を得る。これがnon-collinear Type-I SHGに必要なphase-matching conditionである。

共線の場合には、

$$\gamma=0,\qquad n_{e,\mathrm{eff}}(2\omega,\theta)=n_o(\omega)$$

## 4. BBOの屈折率

BBOのSellmeier式は以下で与えられる。ここで$\lambda$の単位は$\mu\mathrm{m}$である。

$$n_o^2(\lambda)=2.7359+\frac{0.01878}{\lambda^2-0.01822}-0.01354\lambda^2$$

$$n_e^2(\lambda)=2.3753+\frac{0.01224}{\lambda^2-0.01667}-0.01516\lambda^2$$

## 5. Extraordinary SHの有効屈折率

Extraordinary SH wave vectorとoptic axisの間の角度を$\theta$と定義する。このときeffective extraordinary refractive indexは

$$\frac{1}{n_{e,\mathrm{eff}}^2(2\omega,\theta)}=\frac{\cos^2\theta}{n_o^2(2\omega)}+\frac{\sin^2\theta}{n_e^2(2\omega)}$$

で与えられる。

SH beamが$z$方向に伝搬する場合、$\theta=\theta_c$であるため、phase-matching conditionを満たすには

$$\frac{1}{\sqrt{\dfrac{\cos^2\theta}{n_o^2(2\omega)}+\dfrac{\sin^2\theta}{n_e^2(2\omega)}}}=n_o(\omega)\cos\gamma$$

## 6. 現在の条件における角度

上記の条件を満たす結晶内部の角度は

$$\gamma=1.57^\circ$$

であり、空気中の入射角は

$$\gamma'=2.60^\circ$$

を満たせばよい。

焦点距離が $f=200\,\mathrm{mm}$ で、光軸からのbeam offsetが $12.5\,\mathrm{mm}$ の場合、

$$f=200\,\mathrm{mm},\qquad \tan\gamma'=0.0625,\qquad \gamma'=3.58^\circ$$

となる。この角度は厳密には完全なphase-matching angleではないが、差は小さく、結晶も薄いため、ほぼphase-matching条件とみなすことができる。

焦点距離 $75\,\mathrm{mm}$ のlensを使用する場合には、入力beamと光軸の間のoffsetを

$$x=f\tan(2.60^\circ)\simeq 3.4\,\mathrm{mm}\qquad (f=75\,\mathrm{mm})$$

まで狭める必要がある。
