# Introduction

## 1. Nonlinear Polarization

In a dielectric medium, the induced polarization can be expanded in powers of the electric field. In Cartesian tensor notation,

$$
P_i(t)
=
\varepsilon_0
\left[
\sum_j \chi_{ij}^{(1)} E_j(t)
+
\sum_{jk} \chi_{ijk}^{(2)} E_j(t)E_k(t)
+
\sum_{jkl} \chi_{ijkl}^{(3)} E_j(t)E_k(t)E_l(t)
+\cdots
\right].
$$

In compact form,

$$
\mathbf{P}
=
\mathbf{P}^{(1)}
+
\mathbf{P}^{(2)}
+
\mathbf{P}^{(3)}
+\cdots.
$$

The tensor ranks and SI units are as follows.

- $\chi^{(1)}$: rank-2 tensor; dimensionless in SI
- $\chi^{(2)}$: rank-3 tensor; SI unit $\mathrm{m/V}$
- $\chi^{(3)}$: rank-4 tensor; SI unit $\mathrm{m^2/V^2}$

$\mathbf{P}^{(2)}$ is the second-order nonlinear polarization, and  
$\mathbf{P}^{(3)}$ is the third-order nonlinear polarization.

The simple algebraic form above assumes an **instantaneous, local response**. A real dispersive medium is generally described by a response function that depends on the electric field at earlier times; in the frequency domain, this is expressed through a frequency-dependent susceptibility.

A representative scale for estimating the magnitude of optical nonlinearities is the atomic electric field,

$$
E_{\mathrm{at}}
=
\frac{e}{4\pi\varepsilon_0 a_0^2}
\simeq
5.14\times10^{11}\ \mathrm{V/m},
$$

where $a_0$ is the Bohr radius.

The corresponding intensity in vacuum is

$$
I_{\mathrm{at}}
=
\frac{1}{2}c\varepsilon_0 E_{\mathrm{at}}^2
\simeq
3.51\times10^{16}\ \mathrm{W/cm^2}.
$$

If successive terms in the expansion are assumed to become comparable in magnitude when $E\sim E_{\mathrm{at}}$, then

$$
\chi^{(2)}
\sim
\frac{\chi^{(1)}}{E_{\mathrm{at}}},
\qquad
\chi^{(3)}
\sim
\frac{\chi^{(1)}}{E_{\mathrm{at}}^2}.
$$

Taking $\chi^{(1)}\sim1$ gives the approximate estimates

$$
\chi^{(2)}
\sim
2\times10^{-12}\ \mathrm{m/V},
\qquad
\chi^{(3)}
\sim
4\times10^{-24}\ \mathrm{m^2/V^2}.
$$

These values are only order-of-magnitude estimates; the actual values depend strongly on the material and optical frequency.


## 2. Sum- and Difference-Frequency Generation

Consider two monochromatic complex field amplitudes. We use the following convention:

$$
\widetilde{E}(t)
=
E_1 e^{-i\omega_1 t}
+
E_2 e^{-i\omega_2 t}
+
c.c.
$$

First, treating the field as a scalar,

$$
\widetilde{P}^{(2)}(t)
=
\varepsilon_0\chi^{(2)}
\widetilde{E}^{\,2}(t).
$$

Therefore,

$$
\widetilde{P}^{(2)}(t)
=
\varepsilon_0\chi^{(2)}
\left(
E_1 e^{-i\omega_1 t}
+
E_2 e^{-i\omega_2 t}
+
c.c.
\right)^2.
$$

Expanding this expression gives

$$
\widetilde{P}^{(2)}(t)
=
\varepsilon_0\chi^{(2)}
\Big[
E_1^2 e^{-2i\omega_1 t}
+
E_2^2 e^{-2i\omega_2 t}
+
2E_1E_2e^{-i(\omega_1+\omega_2)t}
+
2E_1E_2^*e^{-i(\omega_1-\omega_2)t}
+
c.c.
\Big]
$$

$$
\qquad
+
\varepsilon_0\chi^{(2)}
\left(
E_1E_1^*
+
E_2E_2^*
\right).
$$

This expansion contains second-harmonic, sum-frequency, difference-frequency, and zero-frequency components.

For $\omega_1>\omega_2$, the positive-frequency components are

$$
P^{(2)}(2\omega_1)
=
\varepsilon_0\chi^{(2)}E_1^2,
$$

$$
P^{(2)}(2\omega_2)
=
\varepsilon_0\chi^{(2)}E_2^2,
$$

$$
P^{(2)}(\omega_1+\omega_2)
=
2\varepsilon_0\chi^{(2)}E_1E_2,
$$

$$
P^{(2)}(\omega_1-\omega_2)
=
2\varepsilon_0\chi^{(2)}E_1E_2^*.
$$

The zero-frequency component is optical rectification:

$$
P^{(2)}(0)
=
2\varepsilon_0\chi^{(2)}
\left(
|E_1|^2
+
|E_2|^2
\right).
$$

The complex conjugates of the positive-frequency components form the corresponding negative-frequency components.


## 3. Tensor Representation of Second-Order Frequency Mixing

For an anisotropic medium, the scalar expression above must be written as a tensor contraction. For sum-frequency generation, let

$$
\omega_3=\omega_1+\omega_2.
$$

With the complex-amplitude convention above, the sum-frequency polarization is

$$
P_i^{(2)}(\omega_3)
=
2\varepsilon_0
\sum_{j=x,y,z}
\sum_{k=x,y,z}
\chi_{ijk}^{(2)}
(\omega_3;\omega_1,\omega_2)
E_j(\omega_1)
E_k(\omega_2).
$$


### 3.1 A Useful $3\times3$ Matrix for One Output Polarization

Fix one output component of the polarization. For example, for $i=x$,

$$
\mathbf{\chi}_x^{(2)}
=
\begin{pmatrix}
\chi_{xxx}^{(2)} & \chi_{xxy}^{(2)} & \chi_{xxz}^{(2)}\\
\chi_{xyx}^{(2)} & \chi_{xyy}^{(2)} & \chi_{xyz}^{(2)}\\
\chi_{xzx}^{(2)} & \chi_{xzy}^{(2)} & \chi_{xzz}^{(2)}
\end{pmatrix}.
$$

Then

$$
P_x^{(2)}(\omega_3)
=
2\varepsilon_0
\mathbf{E}^T(\omega_1)
\mathbf{\chi}_x^{(2)}
\mathbf{E}(\omega_2),
$$

where

$$
\mathbf{E}(\omega_m)
=
\begin{pmatrix}
E_x(\omega_m)\\
E_y(\omega_m)\\
E_z(\omega_m)
\end{pmatrix}.
$$

This form is particularly convenient for numerical calculations.

Because the nonlinear polarization for sum-frequency generation contains products of the form $E_jE_k$, the expression uses the ordinary transpose $T$, not the Hermitian transpose $\dagger$.


## 3.2 Intrinsic Permutation Symmetry in SFG with Different Frequencies

Consider the following two terms:

$$
A = E_x(\omega_2)E_y(\omega_1)\chi_{xxy}^{(2)}
$$

and

$$
B = E_y(\omega_2)E_x(\omega_1)\chi_{xyx}^{(2)}.
$$

When the two frequencies are different, i.e., $\omega_1 \neq \omega_2$, in general $A \neq B$.

The reason is that the nonlinear susceptibility depends not only on the Cartesian indices but also on the frequencies associated with those indices.

Writing the frequency arguments of the first term explicitly,

$$
A = \chi_{xxy}^{(2)}(\omega_3;\omega_2,\omega_1)E_x(\omega_2)E_y(\omega_1).
$$

Here, $\omega_3 = \omega_1+\omega_2$.

The intrinsic permutation symmetry of the second-order susceptibility is

$$
\chi_{ijk}^{(2)}(\omega_3;\omega_a,\omega_b)
=
\chi_{ikj}^{(2)}(\omega_3;\omega_b,\omega_a).
$$

Therefore, in the present case,

$$
\chi_{xxy}^{(2)}(\omega_3;\omega_2,\omega_1)
=
\chi_{xyx}^{(2)}(\omega_3;\omega_1,\omega_2).
$$

Thus, intrinsic permutation symmetry does not simply interchange the Cartesian indices $x$ and $y$; rather, it simultaneously interchanges the entire **index-frequency pairs** $(x,\omega_2)$ and $(y,\omega_1)$.

Accordingly, $A$ can be written as

$$
A
=
\chi_{xxy}^{(2)}(\omega_3;\omega_2,\omega_1)
E_x(\omega_2)E_y(\omega_1)
$$

$$
=
\chi_{xyx}^{(2)}(\omega_3;\omega_1,\omega_2)
E_y(\omega_1)E_x(\omega_2).
$$

Hence, the exact relation guaranteed by intrinsic permutation symmetry is

$$
\chi_{xxy}^{(2)}(\omega_3;\omega_2,\omega_1)
E_x(\omega_2)E_y(\omega_1)
=
\chi_{xyx}^{(2)}(\omega_3;\omega_1,\omega_2)
E_y(\omega_1)E_x(\omega_2).
$$

On the other hand,

$$
B
=
\chi_{xyx}^{(2)}(\omega_3;\omega_2,\omega_1)
E_y(\omega_2)E_x(\omega_1)
$$

represents a different frequency-polarization assignment from $A$.

Specifically, for $A$,

- the $\omega_2$ component is $x$-polarized,
- the $\omega_1$ component is $y$-polarized,

whereas for $B$,

- the $\omega_2$ component is $y$-polarized,
- the $\omega_1$ component is $x$-polarized.

Therefore, for general SFG with $\omega_1 \neq \omega_2$, it is possible that $A \neq B$. The key point is that intrinsic permutation symmetry means the simultaneous interchange $(j,\omega_1) \leftrightarrow (k,\omega_2)$, not an independent interchange $j \leftrightarrow k$ alone.

For degenerate SHG, however, $\omega_1=\omega_2=\omega$, so exchanging the frequencies produces no change:

$$
\chi_{ijk}^{(2)}(2\omega;\omega,\omega)
=
\chi_{ikj}^{(2)}(2\omega;\omega,\omega).
$$

Thus, the last two Cartesian indices can be treated symmetrically. This symmetry makes the usual $3\times6$ contracted notation natural for SHG.

Kleinman symmetry can provide additional index symmetries when applicable, but it should not be assumed automatically in a strongly dispersive regime.


## 4. Contracted Notation for SHG

For a symmetric index pair $j,k$, use the following contraction:

$$
xx\rightarrow1,\qquad
yy\rightarrow2,\qquad
zz\rightarrow3,
$$

$$
yz(=zy)\rightarrow4,\qquad
zx(=xz)\rightarrow5,\qquad
xy(=yx)\rightarrow6.
$$

The conventional nonlinear coefficient is defined as

$$
d_{il}
=
\frac{1}{2}\chi_{ijk}^{(2)},
$$

where $l$ represents the pair $jk$.

For a single field at frequency $\omega$,

$$
\mathbf{q}(\mathbf{E})
=
\begin{pmatrix}
E_x^2\\
E_y^2\\
E_z^2\\
2E_yE_z\\
2E_zE_x\\
2E_xE_y
\end{pmatrix}.
$$

Therefore, the SH polarization can be written as

$$
\mathbf{P}^{(2)}(2\omega)
=
2\varepsilon_0
\mathbf{d}\,
\mathbf{q}(\mathbf{E}).
$$


### 4.1 Cross Term between Two Coherent Beams at the Same Frequency

For non-collinear SHG, it is clearer to distinguish the two beams by **beam number**.

Let the positive-frequency spatial amplitude be

$$
\mathbf{E}(\mathbf{r})
=
\mathbf{E}_1(\mathbf{r})
+
\mathbf{E}_2(\mathbf{r}),
$$

with the common time dependence $e^{-i\omega t}$.

Then

$$
\mathbf{q}(\mathbf{E}_1+\mathbf{E}_2)
=
\mathbf{q}(\mathbf{E}_1)
+
\mathbf{q}(\mathbf{E}_2)
+
\mathbf{q}_{12},
$$

where the cross-term vector is

$$
\mathbf{q}_{12}
=
2
\begin{pmatrix}
E_{1x}E_{2x}\\
E_{1y}E_{2y}\\
E_{1z}E_{2z}\\
E_{1y}E_{2z}+E_{1z}E_{2y}\\
E_{1z}E_{2x}+E_{1x}E_{2z}\\
E_{1x}E_{2y}+E_{1y}E_{2x}
\end{pmatrix}.
$$

Therefore,

$$
\mathbf{P}_{12}^{(2)}(2\omega)
=
2\varepsilon_0
\mathbf{d}\,
\mathbf{q}_{12}.
$$

This definition avoids a common factor-of-two ambiguity.

If $\mathbf{q}_{12}$ is instead defined without the leading factor 2, the prefactor in the polarization expression must be written as $4\varepsilon_0$.


## 5. $\beta$-BBO: Crystal Symmetry and the $d$ Matrix

$\beta$-$\mathrm{BaB_2O_4}$ (BBO) is a non-centrosymmetric trigonal crystal with

$$
\text{space group}=R3c,
\qquad
\text{point group}=3m.
$$

In a commonly used crystal-axis convention, the contracted second-order matrix is

$$
\mathbf{d}_{\mathrm{BBO}}
=
\begin{pmatrix}
0 & 0 & 0 & 0 & d_{15} & -d_{22}\\
-d_{22} & d_{22} & 0 & d_{15} & 0 & 0\\
d_{31} & d_{31} & d_{33} & 0 & 0 & 0
\end{pmatrix}.
$$

When Kleinman symmetry is applicable,

$$
d_{15}=d_{31}.
$$

Commonly quoted magnitudes for BBO are approximately

$$
|d_{22}|
\simeq
2.2\ \mathrm{pm/V},
\qquad
|d_{15}|
=
|d_{31}|
\simeq
0.08\ \mathrm{pm/V}.
$$

The signs of the individual $d$ coefficients may depend on the crystallographic-axis convention.

For the matrix convention above, the signs must be used consistently as

$$
d_{21}=-d_{22}.
$$

### 6.1 Derivation of the Cross-Term Nonlinear Polarization for Two $y$-Polarized Beams

Assume that the $(x,y,z)$ axes used above are **exactly the same as the crystal axes** used in Eq. (15).

Let the two fundamental beams have the same frequency $\omega$ and propagate symmetrically in the $x$-$z$ plane at angles $\pm\theta$.

Their electric fields are

$$
\mathbf{E}_1(\mathbf{r})
=
\begin{pmatrix}
0\\
E_y e^{ik_\omega(\cos\theta\,z+\sin\theta\,x)}\\
0
\end{pmatrix},
$$

$$
\mathbf{E}_2(\mathbf{r})
=
\begin{pmatrix}
0\\
E_y e^{ik_\omega(\cos\theta\,z-\sin\theta\,x)}\\
0
\end{pmatrix}.
$$

The total fundamental field is therefore

$$
\mathbf{E}
=
\mathbf{E}_1+\mathbf{E}_2.
$$

Because both beams contain only $y$ polarization,

$$
E_x=0,
\qquad
E_z=0,
$$

and

$$
E_y
=
E_{1y}+E_{2y}.
$$

In the contracted notation for SHG, the quadratic field vector is

$$
\mathbf{q}(\mathbf{E})
=
\begin{pmatrix}
E_x^2\\
E_y^2\\
E_z^2\\
2E_yE_z\\
2E_zE_x\\
2E_xE_y
\end{pmatrix}.
$$

Since $E_x=E_z=0$ in the present case,

$$
\mathbf{q}(\mathbf{E})
=
\begin{pmatrix}
0\\
(E_{1y}+E_{2y})^2\\
0\\
0\\
0\\
0
\end{pmatrix}.
$$

Expanding the second component,

$$
(E_{1y}+E_{2y})^2
=
E_{1y}^2
+
E_{2y}^2
+
2E_{1y}E_{2y}.
$$

Here,

$$
E_{1y}^2
$$

and

$$
E_{2y}^2
$$

are the SHG components generated independently by beams 1 and 2, respectively, while

$$
2E_{1y}E_{2y}
$$

is the **cross-term SHG** between the two beams.

Thus, the contracted field vector corresponding to the cross term is

$$
\mathbf{q}_{12}
=
\begin{pmatrix}
0\\
2E_{1y}E_{2y}\\
0\\
0\\
0\\
0
\end{pmatrix}.
$$

Now substitute the spatial phases of the two fields:

$$
E_{1y}
=
E_y
e^{ik_\omega(\cos\theta\,z+\sin\theta\,x)},
$$

$$
E_{2y}
=
E_y
e^{ik_\omega(\cos\theta\,z-\sin\theta\,x)}.
$$

Their product is

$$
E_{1y}E_{2y}
=
E_y^2
e^{ik_\omega(\cos\theta\,z+\sin\theta\,x)}
e^{ik_\omega(\cos\theta\,z-\sin\theta\,x)}.
$$

Combining the exponents,

$$
E_{1y}E_{2y}
=
E_y^2
e^{ik_\omega[
2\cos\theta\,z
+
\sin\theta\,x
-
\sin\theta\,x
]}.
$$

The $x$-direction phase terms cancel, so

$$
E_{1y}E_{2y}
=
E_y^2
e^{i2k_\omega\cos\theta\,z}.
$$

Therefore, the cross-term vector becomes

$$
\boxed{
\mathbf{q}_{12}
=
2E_y^2
e^{i2k_\omega\cos\theta\,z}
\begin{pmatrix}
0\\
1\\
0\\
0\\
0\\
0
\end{pmatrix}
}.
$$

---

Let the contracted nonlinear coefficient matrix of BBO be

$$
\mathbf{d}
=
\begin{pmatrix}
0 & 0 & 0 & 0 & d_{15} & -d_{22}\\
-d_{22} & d_{22} & 0 & d_{15} & 0 & 0\\
d_{31} & d_{31} & d_{33} & 0 & 0 & 0
\end{pmatrix}.
$$

The SHG nonlinear polarization is

$$
\mathbf{P}^{(2)}
=
2\varepsilon_0
\mathbf{d}\,
\mathbf{q}.
$$

Hence, the cross term is

$$
\mathbf{P}_{12}^{(2)}
=
2\varepsilon_0
\mathbf{d}\,
\mathbf{q}_{12}.
$$

Substituting the expression for $\mathbf{q}_{12}$ obtained above,

$$
\mathbf{P}_{12}^{(2)}
=
2\varepsilon_0
\begin{pmatrix}
0 & 0 & 0 & 0 & d_{15} & -d_{22}\\
-d_{22} & d_{22} & 0 & d_{15} & 0 & 0\\
d_{31} & d_{31} & d_{33} & 0 & 0 & 0
\end{pmatrix}
\left[
2E_y^2
e^{i2k_\omega\cos\theta\,z}
\begin{pmatrix}
0\\
1\\
0\\
0\\
0\\
0
\end{pmatrix}
\right].
$$

Pulling the scalar factor outside,

$$
\mathbf{P}_{12}^{(2)}
=
4\varepsilon_0
E_y^2
e^{i2k_\omega\cos\theta\,z}
\begin{pmatrix}
0 & 0 & 0 & 0 & d_{15} & -d_{22}\\
-d_{22} & d_{22} & 0 & d_{15} & 0 & 0\\
d_{31} & d_{31} & d_{33} & 0 & 0 & 0
\end{pmatrix}
\begin{pmatrix}
0\\
1\\
0\\
0\\
0\\
0
\end{pmatrix}.
$$

Because only the second component of the vector on the right is 1, only the **second column** of the $d$ matrix is selected:

$$
\begin{pmatrix}
0 & 0 & 0 & 0 & d_{15} & -d_{22}\\
-d_{22} & d_{22} & 0 & d_{15} & 0 & 0\\
d_{31} & d_{31} & d_{33} & 0 & 0 & 0
\end{pmatrix}
\begin{pmatrix}
0\\
1\\
0\\
0\\
0\\
0
\end{pmatrix}
=
\begin{pmatrix}
0\\
d_{22}\\
d_{31}
\end{pmatrix}.
$$

Therefore,

$$
\boxed{
\mathbf{P}_{12}^{(2)}
=
4\varepsilon_0
E_y^2
e^{i2k_\omega\cos\theta\,z}
\begin{pmatrix}
0\\
d_{22}\\
d_{31}
\end{pmatrix}
}.
$$

The Cartesian components are

$$
P_{12,x}^{(2)}=0,
$$

$$
P_{12,y}^{(2)}
=
4\varepsilon_0 d_{22}
E_y^2
e^{i2k_\omega\cos\theta\,z},
$$

$$
P_{12,z}^{(2)}
=
4\varepsilon_0 d_{31}
E_y^2
e^{i2k_\omega\cos\theta\,z}.
$$

The spatial phase factor

$$
e^{i2k_\omega\cos\theta\,z}
$$

means that the wave vector of the cross-term nonlinear polarization is

$$
\mathbf{K}_{\mathrm{NL}}
=
\mathbf{k}_1+\mathbf{k}_2
=
\begin{pmatrix}
0\\
0\\
2k_\omega\cos\theta
\end{pmatrix}.
$$

Thus, in the symmetric non-collinear geometry, the spatial phase of the cross-term nonlinear polarization propagates along the $z$ direction, i.e., along the angle bisector of the two fundamental beams.


### 6.2 Calculation Using the Actual Type-I BBO Crystal Coordinates

In Section 6.1, the laboratory $(x,y,z)$ axes were intentionally assumed to be **exactly the same as the crystal axes** used in Eq. (15).  
Therefore, Section 6.1 is retained as a comparison calculation based on that incorrect assumption.

For an actual Type-I BBO crystal, the laboratory and crystal coordinate systems must be distinguished.

Here, the calculation is repeated using the standard $3m$ crystal-axis convention corresponding to the $d$ matrix in Eq. (15).

---

#### 6.2.1 Laboratory Coordinates

Define the laboratory coordinates $(x,y,z)$ as follows.

- $z$: the angle-bisector direction of the two fundamental beams and the propagation direction of the cross-term SH
- $x$: the transverse direction within the plane in which the two beams cross
- $y$: the direction perpendicular to the beam-crossing plane and the polarization direction of the fundamental beams

Thus, the two fundamental wave vectors are

$$
\mathbf{k}_1
=
k_\omega
\begin{pmatrix}
\sin\theta\\
0\\
\cos\theta
\end{pmatrix},
\qquad
\mathbf{k}_2
=
k_\omega
\begin{pmatrix}
-\sin\theta\\
0\\
\cos\theta
\end{pmatrix}.
$$

The spatial wave vector of the cross-term nonlinear polarization is

$$
\mathbf{K}_{\mathrm{NL}}
=
\mathbf{k}_1+\mathbf{k}_2
=
2k_\omega\cos\theta
\begin{pmatrix}
0\\
0\\
1
\end{pmatrix}.
$$

Therefore, the nominal propagation direction of the cross-term SH is the laboratory $z$ direction.

---

#### 6.2.2 BBO Crystal Coordinates $(X,Y,Z)$

We denote the crystal coordinates used in the $d$ matrix of Eq. (15) by uppercase $(X,Y,Z)$.

- $Z$: the crystallographic $c$ axis of BBO and the **optic axis**
- $YZ$ plane: one **crystal mirror plane** of the $3m$ point group
- $X$: the axis perpendicular to that mirror plane

In the standard orientation for Type-I $oo\rightarrow e$ BBO that exploits the large $d_{22}$ coefficient, the propagation plane is placed in the crystal $YZ$ mirror plane.  
In the standard spherical-angle convention, the optimum azimuth can be chosen as

$$
\phi=-\frac{\pi}{2}.
$$

In the present geometry, the laboratory $xz$ plane is the optical principal plane, so we set

$$
\boxed{
YZ_{\mathrm{crys}}
=
xz_{\mathrm{lab}}
}.
$$

Therefore, the crystal $X$ axis is parallel to the laboratory $y$ axis.  
To satisfy both a right-handed coordinate system and the $\phi=-\pi/2$ convention, we define

$$
\boxed{
\hat{\mathbf{X}}
=
-\hat{\mathbf{y}}
}.
$$

This sign choice does not by itself change the physics, but it is necessary to keep the sign of $d_{22}$ and the $\phi$ convention consistent.

Let the generated SH wave vector $\mathbf{k}_{2\omega}$ point along the laboratory $z$ direction, and let $\alpha$ be the angle between the optic axis $Z$ and the $z$ axis.

The crystal basis vectors expressed in laboratory coordinates can then be written as

$$
\boxed{
\hat{\mathbf{X}}
=
\begin{pmatrix}
0\\
-1\\
0
\end{pmatrix},
\qquad
\hat{\mathbf{Y}}
=
\begin{pmatrix}
\cos\alpha\\
0\\
-\sin\alpha
\end{pmatrix},
\qquad
\hat{\mathbf{Z}}
=
\begin{pmatrix}
\sin\alpha\\
0\\
\cos\alpha
\end{pmatrix}
}.
$$

Indeed,

$$
\hat{\mathbf{X}}\times\hat{\mathbf{Y}}
=
\hat{\mathbf{Z}},
$$

so the coordinate system is right-handed.

The rotation matrix that transforms crystal-frame components into the laboratory frame has these crystal basis vectors as its columns:

$$
\boxed{
R
=
\begin{pmatrix}
0 & \cos\alpha & \sin\alpha\\
-1 & 0 & 0\\
0 & -\sin\alpha & \cos\alpha
\end{pmatrix}
}.
$$

Therefore,

$$
\boxed{
\mathbf{v}_{\mathrm{lab}}
=
R\mathbf{v}_{\mathrm{crys}},
\qquad
\mathbf{v}_{\mathrm{crys}}
=
R^T\mathbf{v}_{\mathrm{lab}}.
}
$$

---

#### 6.2.3 Transforming the Wave Vectors into the Crystal Frame

First, transform the generated SH direction:

$$
R^T
\begin{pmatrix}
0\\
0\\
1
\end{pmatrix}
=
\begin{pmatrix}
0\\
-\sin\alpha\\
\cos\alpha
\end{pmatrix}.
$$

Thus, in crystal coordinates,

$$
\boxed{
\hat{\mathbf{k}}_{2\omega,\mathrm{crys}}
=
\begin{pmatrix}
0\\
-\sin\alpha\\
\cos\alpha
\end{pmatrix}
}.
$$

Therefore, $\mathbf{k}_{2\omega}$ lies in the crystal $YZ$ mirror plane and makes an angle $\alpha$ with the $Z$ axis.  
Because its $X$ component is zero and $Y<0$, this orientation corresponds to

$$
\phi=-90^\circ.
$$

For the two fundamental beams,

$$
R^T\mathbf{k}_1
=
k_\omega
\begin{pmatrix}
0\\
-\sin(\alpha-\theta)\\
\cos(\alpha-\theta)
\end{pmatrix},
$$

$$
R^T\mathbf{k}_2
=
k_\omega
\begin{pmatrix}
0\\
-\sin(\alpha+\theta)\\
\cos(\alpha+\theta)
\end{pmatrix}.
$$

Hence, $\mathbf{k}_1$, $\mathbf{k}_2$, $\mathbf{k}_{2\omega}$, and the optic axis $Z$ all lie in the crystal $YZ$ mirror plane.

---

#### 6.2.4 The Fundamental Polarization Is the Ordinary Polarization along Crystal $X$

In the laboratory frame, the two fundamental fields are

$$
\mathbf{E}_{1,\mathrm{lab}}
=
E_y
e^{ik_\omega(\cos\theta\,z+\sin\theta\,x)}
\begin{pmatrix}
0\\
1\\
0
\end{pmatrix},
$$

$$
\mathbf{E}_{2,\mathrm{lab}}
=
E_y
e^{ik_\omega(\cos\theta\,z-\sin\theta\,x)}
\begin{pmatrix}
0\\
1\\
0
\end{pmatrix}.
$$

Transforming to the crystal frame,

$$
\mathbf{E}_{m,\mathrm{crys}}
=
R^T\mathbf{E}_{m,\mathrm{lab}}.
$$

Since

$$
R^T
\begin{pmatrix}
0\\
1\\
0
\end{pmatrix}
=
\begin{pmatrix}
-1\\
0\\
0
\end{pmatrix},
$$

we obtain

$$
\boxed{
\mathbf{E}_{1,\mathrm{crys}}
=
-E_y
e^{ik_\omega(\cos\theta\,z+\sin\theta\,x)}
\begin{pmatrix}
1\\
0\\
0
\end{pmatrix}
}
$$

and

$$
\boxed{
\mathbf{E}_{2,\mathrm{crys}}
=
-E_y
e^{ik_\omega(\cos\theta\,z-\sin\theta\,x)}
\begin{pmatrix}
1\\
0\\
0
\end{pmatrix}.
}
$$

Thus, although the beams are $y$-polarized in the laboratory frame, they are **$X$-polarized in the crystal coordinates** used in Eq. (15):

$$
\boxed{
E_{X,\mathrm{crys}}=-E_y,
\qquad
E_{Y,\mathrm{crys}}=E_{Z,\mathrm{crys}}=0.
}
$$

The minus sign disappears in the SHG quantity $E_X^2$.

The ordinary polarization is perpendicular to the principal plane formed by $\mathbf{k}$ and the optic axis $Z$.  
Because the principal plane here is the crystal $YZ$ plane, the ordinary polarization is exactly along the crystal $X$ direction.

This is the most important difference from the incorrect coordinate identification used in Section 6.1.

---

#### 6.2.5 Cross-Term Contracted Vector in the Crystal Frame

The product of the $X$ components of the two fields is

$$
E_{1X}E_{2X}
=
(-E_y)(-E_y)
e^{ik_\omega(\cos\theta\,z+\sin\theta\,x)}
e^{ik_\omega(\cos\theta\,z-\sin\theta\,x)}.
$$

Therefore,

$$
E_{1X}E_{2X}
=
E_y^2
e^{i2k_\omega\cos\theta\,z}.
$$

The cross term in the square of the total field is

$$
(E_{1X}+E_{2X})^2
\supset
2E_{1X}E_{2X},
$$

so

$$
\boxed{
\mathbf{q}_{12,\mathrm{crys}}
=
2E_y^2
e^{i2k_\omega\cos\theta\,z}
\begin{pmatrix}
1\\
0\\
0\\
0\\
0\\
0
\end{pmatrix}.
}
$$

Thus, in the actual Type-I orientation, the **first column**, not the second column, of the $d$ matrix is selected.

---

#### 6.2.6 Applying the BBO $d$ Matrix

The BBO matrix in Eq. (15) is

$$
\mathbf{d}_{\mathrm{BBO}}
=
\begin{pmatrix}
0 & 0 & 0 & 0 & d_{15} & -d_{22}\\
-d_{22} & d_{22} & 0 & d_{15} & 0 & 0\\
d_{31} & d_{31} & d_{33} & 0 & 0 & 0
\end{pmatrix}.
$$

Therefore, the cross-term nonlinear polarization is

$$
\mathbf{P}_{12,\mathrm{crys}}^{(2)}
=
2\varepsilon_0
\mathbf{d}_{\mathrm{BBO}}
\mathbf{q}_{12,\mathrm{crys}}.
$$

Thus,

$$
\mathbf{P}_{12,\mathrm{crys}}^{(2)}
=
4\varepsilon_0
E_y^2
e^{i2k_\omega\cos\theta\,z}
\begin{pmatrix}
0 & 0 & 0 & 0 & d_{15} & -d_{22}\\
-d_{22} & d_{22} & 0 & d_{15} & 0 & 0\\
d_{31} & d_{31} & d_{33} & 0 & 0 & 0
\end{pmatrix}
\begin{pmatrix}
1\\
0\\
0\\
0\\
0\\
0
\end{pmatrix}.
$$

Only the first column remains:

$$
\boxed{
\mathbf{P}_{12,\mathrm{crys}}^{(2)}
=
4\varepsilon_0
E_y^2
e^{i2k_\omega\cos\theta\,z}
\begin{pmatrix}
0\\
-d_{22}\\
d_{31}
\end{pmatrix}.
}
$$

Therefore, in the crystal frame,

$$
P_X^{(2)}=0,
$$

$$
P_Y^{(2)}
=
-4\varepsilon_0 d_{22}
E_y^2
e^{i2k_\omega\cos\theta\,z},
$$

$$
P_Z^{(2)}
=
4\varepsilon_0 d_{31}
E_y^2
e^{i2k_\omega\cos\theta\,z}.
$$

---

#### 6.2.7 Rotating the Nonlinear Polarization Back to the Laboratory Frame

In the laboratory frame,

$$
\mathbf{P}_{12,\mathrm{lab}}^{(2)}
=
R
\mathbf{P}_{12,\mathrm{crys}}^{(2)}.
$$

Therefore,

$$
\mathbf{P}_{12,\mathrm{lab}}^{(2)}
=
4\varepsilon_0
E_y^2
e^{i2k_\omega\cos\theta\,z}
\begin{pmatrix}
0 & \cos\alpha & \sin\alpha\\
-1 & 0 & 0\\
0 & -\sin\alpha & \cos\alpha
\end{pmatrix}
\begin{pmatrix}
0\\
-d_{22}\\
d_{31}
\end{pmatrix}.
$$

Carrying out the matrix multiplication,

$$
\boxed{
\mathbf{P}_{12,\mathrm{lab}}^{(2)}
=
4\varepsilon_0
E_y^2
e^{i2k_\omega\cos\theta\,z}
\begin{pmatrix}
-d_{22}\cos\alpha+d_{31}\sin\alpha\\
0\\
d_{22}\sin\alpha+d_{31}\cos\alpha
\end{pmatrix}.
}
$$

Thus,

$$
\boxed{
P_x^{(2)}
=
4\varepsilon_0
\left(
d_{31}\sin\alpha-d_{22}\cos\alpha
\right)
E_y^2
e^{i2k_\omega\cos\theta\,z}
}
$$

and

$$
P_y^{(2)}=0,
$$

$$
P_z^{(2)}
=
4\varepsilon_0
\left(
d_{22}\sin\alpha+d_{31}\cos\alpha
\right)
E_y^2
e^{i2k_\omega\cos\theta\,z}.
$$

---

#### 6.2.8 Effective Nonlinear Coefficient Coupled to the Extraordinary SH

The generated SH propagates along the laboratory $z$ direction, while the optic axis lies in the laboratory $xz$ plane.

Therefore, the extraordinary output polarization basis for Type-I $oo\rightarrow e$ can be taken along the laboratory $x$ direction:

$$
\hat{\mathbf{e}}_{2\omega,e}^{\mathrm{lab}}
=
\begin{pmatrix}
1\\
0\\
0
\end{pmatrix}.
$$

Transforming this vector into crystal coordinates,

$$
\hat{\mathbf{e}}_{2\omega,e}^{\mathrm{crys}}
=
R^T
\hat{\mathbf{e}}_{2\omega,e}^{\mathrm{lab}}
=
\begin{pmatrix}
0\\
\cos\alpha\\
\sin\alpha
\end{pmatrix}.
$$

The projection onto the crystal-frame nonlinear polarization therefore gives

$$
d_{\mathrm{eff}}
=
\begin{pmatrix}
0&
\cos\alpha&
\sin\alpha
\end{pmatrix}
\begin{pmatrix}
0\\
-d_{22}\\
d_{31}
\end{pmatrix}.
$$

Hence,

$$
\boxed{
d_{\mathrm{eff}}^{ooe}
=
d_{31}\sin\alpha
-
d_{22}\cos\alpha.
}
$$

This agrees with the general Type-I expression for a $3m$ crystal,

$$
d_{\mathrm{eff}}^{ooe}
=
d_{31}\sin\alpha
-
d_{22}\cos\alpha\sin3\phi,
$$

when the standard BBO convention

$$
\phi=-90^\circ
$$

is applied.

Note that different references may use different signs for the $X,Y$ axes and for $d_{22}$. Therefore, maintaining a **single consistent axis convention throughout the calculation** is more important than the overall sign of the expression.

If the spatial walk-off angle $\rho$ is included in a more precise calculation, the standard expression is approximately

$$
d_{\mathrm{eff}}^{ooe}
=
d_{31}\sin(\alpha+\rho)
-
d_{22}\cos(\alpha+\rho).
$$

---

#### 6.2.9 Caution When Using Numerical Values of $d_{22}$ and $d_{31}$

Values given by manufacturers or data sheets, such as

$$
|d_{22}|
\simeq
2.2\ \mathrm{pm/V},
\qquad
|d_{31}|
\simeq
0.08\ \mathrm{pm/V},
$$

should often be interpreted as **magnitudes**.

Under a standard BBO axis convention, the relative signs of $d_{22}$ and $d_{31}$ must be used consistently.

For example, choosing

$$
d_{31}=+0.08\ \mathrm{pm/V},
\qquad
d_{22}=-2.2\ \mathrm{pm/V},
$$

gives

$$
d_{\mathrm{eff}}
=
0.08\sin\alpha
+
2.2\cos\alpha.
$$

For $\alpha=29.2^\circ$,

$$
\boxed{
d_{\mathrm{eff}}
\simeq
1.96\ \mathrm{pm/V}.
}
$$

If all crystal-axis sign conventions are reversed, the overall sign of $d_{\mathrm{eff}}$ may change. However, the SHG conversion efficiency depends on

$$
I_{2\omega}\propto |d_{\mathrm{eff}}|^2,
$$

so the observable intensity is unchanged.

---

#### 6.2.10 Difference from Section 6.1

Section 6.1 used the incorrect assumption

$$
y_{\mathrm{lab}}
\equiv
Y_{\mathrm{crys}},
$$

which selected the second column of the $d$ matrix and yielded

$$
\mathbf{P}^{(2)}
\propto
\begin{pmatrix}
0\\
d_{22}\\
d_{31}
\end{pmatrix}.
$$

However, in the standard orientation of an actual Type-I BBO crystal,

$$
\boxed{
y_{\mathrm{lab}}
\parallel
X_{\mathrm{crys}}
},
$$

and, for the present right-handed $\phi=-90^\circ$ convention,

$$
\boxed{
\hat{\mathbf{X}}
=
-\hat{\mathbf{y}}.
}
$$

Therefore, the actual cross term uses the first column of the $d$ matrix:

$$
\boxed{
\mathbf{P}_{12,\mathrm{crys}}^{(2)}
\propto
\begin{pmatrix}
0\\
-d_{22}\\
d_{31}
\end{pmatrix}.
}
$$

After rotation to the laboratory frame and projection onto the extraordinary output,

$$
\boxed{
d_{\mathrm{eff}}^{ooe}
=
d_{31}\sin\alpha
-
d_{22}\cos\alpha
}
$$

is obtained.

Thus, the essential error in Section 6.1 is not only that the tilt of the optic axis was ignored, but also that the **laboratory $y$ polarization was incorrectly interpreted as the crystal $Y$ component**.

For the standard Type-I BBO orientation, the laboratory $y$ polarization corresponds to the ordinary polarization along crystal $X$.
