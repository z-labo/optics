## Wave Equation for Nonlinear Optical Media

### 1–2. From Maxwell's equations to the general wave equation

Derive the wave equation by inserting Ampère's law into Faraday's law:

$$
\vec{\nabla}\times\vec{\nabla}\times\widetilde{\vec E}
=
-\mu_0\partial_t^2\widetilde{\vec D}
$$

$$
\widetilde{\vec D}
=
\varepsilon_0\widetilde{\vec E}
+
\widetilde{\vec P},
\qquad
\mu_0=\frac{1}{\varepsilon_0c^2}
$$

$$
\vec{\nabla}\times\vec{\nabla}\times\widetilde{\vec E}
+
\frac{1}{c^2}\partial_t^2\widetilde{\vec E}
=
-\frac{1}{\varepsilon_0c^2}\partial_t^2\widetilde{\vec P}
\tag{5}
$$


Equation (5) is the most general form of the wave equation used here in classical nonlinear optics.

### 3–4. Plane-wave condition and separation of polarization

In optics, we would like to replace
$\vec{\nabla}\times\vec{\nabla}\times\widetilde{\vec E}$
with a Laplacian. In general, however, the divergence of the electric field is not zero. For a plane wave we use

$$
\nabla\cdot\vec E=0.
$$

This leads to

$$
-\nabla^2\widetilde{\vec E}
+
\frac{1}{c^2}\partial_t^2\widetilde{\vec E}
=
-\frac{1}{\varepsilon_0c^2}\partial_t^2\widetilde{\vec P}
\tag{6}
$$

$$
\widetilde{\vec P}
=
\widetilde{\vec P}^{(1)}
+
\widetilde{\vec P}^{NL}
\tag{7}
$$

$$
-\nabla^2\widetilde{\vec E}
+
\frac{1}{c^2}\partial_t^2
\left(
\widetilde{\vec E}
+
\frac{1}{\varepsilon_0}\widetilde{\vec P}^{(1)}
\right)
=
-\frac{1}{\varepsilon_0c^2}
\partial_t^2\widetilde{\vec P}^{NL}
\tag{8}
$$

Using

$$
\widetilde{\vec D}^{(1)}
=
\varepsilon_0\widetilde{\vec E}
+
\widetilde{\vec P}^{(1)},
$$

we obtain

$$
-\nabla^2\widetilde{\vec E}
+
\frac{1}{\varepsilon_0c^2}
\partial_t^2\widetilde{\vec D}^{(1)}
=
-\frac{1}{\varepsilon_0c^2}
\partial_t^2\widetilde{\vec P}^{NL}.
\tag{9}
$$

### 5–6. Simple case: lossless, dispersionless medium

For a lossless, dispersionless medium, assume an instantaneous linear response:

$$
\widetilde{\vec D}^{(1)}(t)
=
\varepsilon_0\varepsilon\,
\widetilde{\vec E}(t)
\tag{10}
$$

Substituting Eq. (10) into Eq. (9),
$$
-\nabla^2\widetilde{\vec E}
+
\frac{\varepsilon^{(1)}}{c^2}
\partial_t^2\widetilde{\vec E}
=
-\frac{1}{\varepsilon_0c^2}
\partial_t^2\widetilde{\vec P}^{NL}.
\tag{11}
$$

In general,
$$
D(t)\neq E(t),
\qquad
D(\omega)\sim E(\omega).
$$


Equation (11) is a wave equation with a source term: the nonlinear-polarization term on its right-hand side.

### 7–11. General case: dispersive media and monochromatic components

For a dispersive medium, Eq. (10) does not hold in the time domain. In frequency space, however, the linear displacement and electric field are related by the frequency-dependent dielectric response.

A second-order nonlinear process couples three waves, while a third-order process couples four waves. Since only a few waves are involved, the total field can be written as a sum of monochromatic components.

$$
\widetilde{\vec D}^{(1)}(\omega)
=
\varepsilon_0\varepsilon^{(1)}(\omega)
\widetilde{\vec E}(\omega)
$$

$$
\widetilde{\vec E}(\vec r,t)
=
\sum_n' \widetilde{\vec E}_n(\vec r,t)
$$

$$
\widetilde{\vec D}(\vec r,t)
=
\sum_n' \widetilde{\vec D}_n(\vec r,t)
$$

$$
\widetilde{\vec P}^{NL}(\vec r,t)
=
\sum_n' \widetilde{\vec P}^{NL}_n(\vec r,t)
\tag{13}
$$

Assuming $\varepsilon^{(1)}\in\mathbb{R}$ and substituting the frequency component into Eq. (9),

$$
-\nabla^2\widetilde{\vec E}_n
+
\frac{\varepsilon^{(1)}(\omega_n)}{c^2}
\partial_t^2\widetilde{\vec E}_n
=
-\frac{1}{\varepsilon_0c^2}
\partial_t^2\widetilde{\vec P}^{NL}_n.
\tag{15}
$$

For monochromatic waves,

$$
\widetilde{\vec E}_n(\vec r,t)
=
\vec E_n(\vec r)e^{-i\omega_nt}
+c.c.,
\qquad
\vec E_n\in\mathbb C^3,
$$

$$
\widetilde{\vec D}_n(\vec r,t)
=
\vec D_n(\vec r)e^{-i\omega_nt}
+c.c.,
\qquad
\vec D_n\in\mathbb C^3,
$$

$$
\widetilde{\vec P}_n(\vec r,t)
=
\vec P_n(\vec r)e^{-i\omega_nt}
+c.c.,
\qquad
\vec P_n\in\mathbb C^3.
\tag{16}
$$

Because the spatial and temporal dependences have been factorized,

$$
\partial_t^2\sim-\omega^2,
$$

and therefore
$$
\nabla^2\widetilde{\vec E}_n(\vec r)
+
\frac{\varepsilon^{(1)}(\omega_n)\omega_n^2}{c^2}
\widetilde{\vec E}_n
=
-\frac{\omega_n^2}{\varepsilon_0c^2}
\widetilde{\vec P}^{NL}_n.
\tag{17}
$$


Equation (15) has the same algebraic form as Eq. (11), while allowing the material response to depend on frequency.

## Coupled-Wave Equations for Sum-Frequency Generation (SFG)

### 1–4. Field equation and slowly changing amplitude ansatz

We now apply the preceding wave equation to a specific nonlinear interaction: sum-frequency generation (SFG).

For SFG,

$$
\omega_3=\omega_1+\omega_2.
$$

Equation (15) for the $\omega_3$ component becomes

$$
-\nabla^2\widetilde{\vec E}_3
+
\frac{n_3^2}{c^2}
\partial_t^2\widetilde{\vec E}_3
=
-\frac{1}{\varepsilon_0c^2}
\partial_t^2\widetilde{\vec P}^{NL}_3.
\tag{20}
$$

If the right-hand side is negligible, the ordinary wave-equation solution is

$$
\widetilde E_3(z,t)
=
A_3e^{i(k_3z-\omega_3t)}
+c.c.,
$$

$$
k_3=\frac{n_3\omega_3}{c}.
\tag{21}
$$

When the nonlinear source is small but nonzero, we use the ansatz

$$
\widetilde E_3(z,t)
=
A_3(z)e^{i(k_3z-\omega_3t)}
+c.c.
$$


The $z$-dependent amplitude is physically natural because the generated wave is expected to grow as it propagates through the crystal.

### 5–6. Nonlinear source polarization

For the nonlinear source,

$$
\widetilde P_3(z,t)
=
P_3(z)e^{-i\omega_3t}
+c.c.,
$$

$$
P_3
=
4\varepsilon_0d_{\mathrm{eff}}E_1E_2,
\qquad
2d_{\mathrm{eff}}\sim\chi^{(2)}.
\tag{24}
$$

Also,

$$
E_i(z,t)
=
E_ie^{-i\omega_it}+c.c.,
\qquad
E_i=A_ie^{ik_iz}.
\tag{25}
$$

Thus,

$$
P_3
=
4\varepsilon_0d_{\mathrm{eff}}
A_1A_2e^{i(k_1+k_2)z}.
\tag{26}
$$

### 7. Substitute into the wave equation

Substitute the field ansatz and Eq. (26) into Eq. (20).

First term:

$$
-\nabla^2\widetilde{\vec E}_3
=
-\partial_z^2
\left[
A_3(z)e^{i(k_3z-\omega_3t)}+c.c.
\right]
$$

$$
=
-
\left[
\frac{\partial^2A_3}{\partial z^2}
+
2ik_3\frac{\partial A_3}{\partial z}
-
k_3^2A_3
\right]
e^{i(k_3z-\omega_3t)}
+c.c.
$$

Second term:

$$
\frac{n_3^2}{c^2}\partial_t^2\widetilde{\vec E}_3
=
-\frac{n_3^2\omega_3^2}{c^2}
\left[
A_3e^{i(k_3z-\omega_3t)}+c.c.
\right].
$$

Third term:

$$
-\frac{1}{\varepsilon_0c^2}
\partial_t^2\widetilde{\vec P}^{NL}_3
=
\frac{4\varepsilon_0d_{\mathrm{eff}}\omega_3^2}
{\varepsilon_0c^2}
A_1A_2
e^{i[(k_1+k_2)z-\omega_3t]}
+c.c.
$$

### 10. Collect the terms

Combining the terms,

$$
\left[
-\frac{\partial^2A_3}{\partial z^2}
-2ik_3\frac{\partial A_3}{\partial z}
+k_3^2A_3
-\frac{n_3^2\omega_3^2}{c^2}A_3
\right]
e^{i(k_3z-\omega_3t)}
+c.c.
=
\frac{4\varepsilon_0d_{\mathrm{eff}}\omega_3^2}
{\varepsilon_0c^2}
A_1A_2
e^{i[(k_1+k_2)z-\omega_3t]}
+c.c.
$$

Since

$$
k_3^2A_3-\frac{n_3^2\omega_3^2}{c^2}A_3=0,
$$
and because $e^{-i\omega t}$ and $e^{i\omega t}$ are linearly independent, the $c.c.$ term can be treated separately. Dividing by $e^{-i\omega_3t}$ gives
$$
-
\left[
\frac{\partial^2A_3}{\partial z^2}
+
2ik_3\frac{\partial A_3}{\partial z}
\right]
e^{ik_3z}
=
\frac{4\varepsilon_0d_{\mathrm{eff}}\omega_3^2}
{\varepsilon_0c^2}
A_1A_2e^{i(k_1+k_2)z}.
\tag{27}
$$

### 11–12. Slowly varying amplitude approximation (SVAA)

Under the slowly varying amplitude approximation (SVAA), the second derivative is neglected:

$$
\frac{\partial^2A_3}{\partial z^2}
\ll
k_3\frac{\partial A_3}{\partial z}.
$$

Then

$$
-2ik_3\frac{\partial A_3}{\partial z}e^{ik_3z}
=
\frac{4\varepsilon_0d_{\mathrm{eff}}\omega_3^2}
{\varepsilon_0c^2}
A_1A_2e^{i(k_1+k_2)z},
$$

so that

$$
\frac{\partial A_3}{\partial z}
=
i\frac{2d_{\mathrm{eff}}\omega_3^2}{k_3c^2}
A_1A_2e^{i(k_1+k_2-k_3)z}.
$$

Define

$$
\Delta k=k_1+k_2-k_3,
\qquad
k_3=\frac{n_3\omega_3}{c}.
$$

Therefore,

$$
\frac{\partial A_3}{\partial z}
=
i\frac{2d_{\mathrm{eff}}\omega_3}{n_3c}
A_1A_2e^{i\Delta kz}.
\tag{28}
$$

### 13. Coupled equations for all three waves

The same procedure is applied to $A_1$ and $A_2$, using

$$
\omega_1=\omega_3-\omega_2,
\qquad
\omega_2=\omega_3-\omega_1.
$$

The nonlinear polarizations are

$$
P_1
=
4\varepsilon_0d_{\mathrm{eff}}
A_3A_2^*e^{i(k_3-k_2)z},
$$

$$
P_2
=
4\varepsilon_0d_{\mathrm{eff}}
A_3A_1^*e^{i(k_3-k_1)z}.
$$

Therefore,
<div style="border: 4px solid; padding: 18px 24px; margin: 20px 0; border-radius: 6px;">

$$
\begin{aligned}
\frac{\partial A_1}{\partial z}
&=
i\frac{2d_{\mathrm{eff}}\omega_1^2}{k_1c^2}
A_3(z)A_2^*(z)e^{-i\Delta kz},
\\[8pt]
\frac{\partial A_2}{\partial z}
&=
i\frac{2d_{\mathrm{eff}}\omega_2^2}{k_2c^2}
A_3(z)A_1^*(z)e^{-i\Delta kz},
\\[8pt]
\frac{\partial A_3}{\partial z}
&=
i\frac{2d_{\mathrm{eff}}\omega_3^2}{k_3c^2}
A_1(z)A_2(z)e^{i\Delta kz}.
\end{aligned}
$$

</div>

## 3.4 Phase (Mis)match Considerations

### 1. Undepleted-pump approximation

We assume that $A_3$ remains small compared with $A_1$ and $A_2$. Therefore, $A_1$ and $A_2$ are taken to be independent of $z$:

$$
\frac{\partial A_3}{\partial z}
=
i\frac{2d_{\mathrm{eff}}\omega_3^2}{k_3c^2}
A_1A_2e^{i\Delta kz}.
$$

### 2. Perfect phase matching: $\Delta k=0$

To make things even simpler, assume perfect phase matching:

$$
\Delta k=0.
$$

Then $A_3(z)$ increases linearly with $z$. Therefore, the intensity $I_3(z)$ increases quadratically:

$$
A_3(z)\propto z,
\qquad
I_3(z)\propto z^2.
$$

### 3. Phase mismatch: $\Delta k\neq0$

Now consider

$$
\Delta k\neq0.
$$

Integrating from $z=0$ to $z=L$,

$$
A_3(L)
=
i\frac{2d_{\mathrm{eff}}\omega_3^2}{k_3c^2}
A_1A_2
\int_0^L e^{i\Delta kz}\,dz,
$$

so that

$$
A_3(L)
=
i\frac{2d_{\mathrm{eff}}\omega_3^2}{k_3c^2}
A_1A_2
\frac{e^{i\Delta kL}-1}{i\Delta k}.
$$

### 4. Intensity of the generated wave

To see the effect of $\Delta k\neq0$, calculate $I_3$:

$$
I_3
=
\frac{cn_3\varepsilon_0}{2}|A_3|^2
=
\frac{2\varepsilon_0d_{\mathrm{eff}}^2\omega_3^2}{n_3c^2}
|A_1|^2|A_2|^2
\left|
\frac{e^{i\Delta kL}-1}{\Delta k}
\right|^2.
$$

Also,

$$
c\varepsilon_0=\frac{1}{377\ \Omega}.
$$

### 5. Sinc dependence

Using

$$
\left|
\frac{e^{i\Delta kL}-1}{\Delta k}
\right|^2
=
L^2
\operatorname{sinc}^2
\left(\frac{\Delta kL}{2}\right),
$$

the phase-mismatch dependence is expressed by the familiar $\operatorname{sinc}^2$ function.

### 6. Expressing $I_3$ in terms of $I_1$ and $I_2$

Using

$$
I_1=\frac{cn_1\varepsilon_0}{2}|A_1|^2,
\qquad
I_2=\frac{cn_2\varepsilon_0}{2}|A_2|^2,
$$

we obtain

<div style="border: 4px solid; padding: 18px 24px; margin: 20px 0; border-radius: 6px;">

$$
I_3
=
8
\frac{
d_{\mathrm{eff}}^2\omega_3^2 I_1 I_2
}{
n_1n_2n_3\varepsilon_0c^3
}
L^2
\operatorname{sinc}^2
\left(
\frac{\Delta kL}{2}
\right).
$$

</div>

Following the supplied notes,

$$
L\equiv L_{\mathrm{coh}}=\frac{2}{\Delta k}.
$$
