# Susceptibility in NLC

## 1. 非線形分極

誘電体媒質において、誘起分極は電場のべき級数として展開できる。Cartesian tensor notationで書くと、

$$
P_i(t)
=
\varepsilon_0
\left[
\sum_j \chi_{ij}^{(1)} E_j(t)
+
\sum_{jk} \chi_{ijk}^{(2)} E_j(t)E_k(t)
+
\sum_{jkl} \chi_{ijkl}^{(3)} E_j(t)E_k(t)E_l(t)
+\cdots
\right].
$$

簡略化して書くと、

$$
\mathbf{P}
=
\mathbf{P}^{(1)}
+
\mathbf{P}^{(2)}
+
\mathbf{P}^{(3)}
+\cdots.
$$

tensor rankとSI単位は以下のとおりである。

- $\chi^{(1)}$: rank-2 tensor、SIでは無次元
- $\chi^{(2)}$: rank-3 tensor、SI単位は $\mathrm{m/V}$
- $\chi^{(3)}$: rank-4 tensor、SI単位は $\mathrm{m^2/V^2}$

$\mathbf{P}^{(2)}$は2次非線形分極（second-order nonlinear polarization）、  
$\mathbf{P}^{(3)}$は3次非線形分極（third-order nonlinear polarization）である。

上記の単純な代数的表現は **instantaneous, local response** を仮定している。実際の分散媒質は一般に、過去の時刻における電場に依存するresponse functionで記述され、周波数領域ではfrequency-dependent susceptibilityとして表される。

非線形性の大きさを概算するための代表的なスケールとしてatomic electric fieldがある。

$$
E_{\mathrm{at}}
=
\frac{e}{4\pi\varepsilon_0 a_0^2}
\simeq
5.14\times10^{11}\ \mathrm{V/m},
$$

ここで $a_0$ はBohr radiusである。

これに対応する真空中のintensityは

$$
I_{\mathrm{at}}
=
\frac{1}{2}c\varepsilon_0 E_{\mathrm{at}}^2
\simeq
3.51\times10^{16}\ \mathrm{W/cm^2}.
$$

$E\sim E_{\mathrm{at}}$ において展開式の連続する各項が同程度の大きさになると仮定すると、

$$
\chi^{(2)}
\sim
\frac{\chi^{(1)}}{E_{\mathrm{at}}},
\qquad
\chi^{(3)}
\sim
\frac{\chi^{(1)}}{E_{\mathrm{at}}^2}.
$$

$\chi^{(1)}\sim1$ とすると、おおよそ

$$
\chi^{(2)}
\sim
2\times10^{-12}\ \mathrm{m/V},
\qquad
\chi^{(3)}
\sim
4\times10^{-24}\ \mathrm{m^2/V^2}.
$$

これらはorder-of-magnitude estimateにすぎず、実際の値は物質および光学周波数に強く依存する。


## 2. 和周波・差周波発生

2つの単色波のcomplex field amplitudeを考える。以下のconventionを用いる。

$$
\widetilde{E}(t)
=
E_1 e^{-i\omega_1 t}
+
E_2 e^{-i\omega_2 t}
+
c.c.
$$

まずfieldをscalarとして扱うと、

$$
\widetilde{P}^{(2)}(t)
=
\varepsilon_0\chi^{(2)}
\widetilde{E}^{\,2}(t).
$$

したがって、

$$
\widetilde{P}^{(2)}(t)
=
\varepsilon_0\chi^{(2)}
\left(
E_1 e^{-i\omega_1 t}
+
E_2 e^{-i\omega_2 t}
+
c.c.
\right)^2.
$$

これを展開すると、

$$
\widetilde{P}^{(2)}(t)
=
\varepsilon_0\chi^{(2)}
\Big[
E_1^2 e^{-2i\omega_1 t}
+
E_2^2 e^{-2i\omega_2 t}
+
2E_1E_2e^{-i(\omega_1+\omega_2)t}
+
2E_1E_2^*e^{-i(\omega_1-\omega_2)t}
+
c.c.
\Big]
$$

$$
\qquad
+
\varepsilon_0\chi^{(2)}
\left(
E_1E_1^*
+
E_2E_2^*
\right).
$$

この展開にはsecond-harmonic、sum-frequency、difference-frequency、およびzero-frequency成分が含まれる。

$\omega_1>\omega_2$ のとき、positive-frequency成分は

$$
P^{(2)}(2\omega_1)
=
\varepsilon_0\chi^{(2)}E_1^2,
$$

$$
P^{(2)}(2\omega_2)
=
\varepsilon_0\chi^{(2)}E_2^2,
$$

$$
P^{(2)}(\omega_1+\omega_2)
=
2\varepsilon_0\chi^{(2)}E_1E_2,
$$

$$
P^{(2)}(\omega_1-\omega_2)
=
2\varepsilon_0\chi^{(2)}E_1E_2^*.
$$

zero-frequency成分は以下に示すoptical rectificationである。

$$
P^{(2)}(0)
=
2\varepsilon_0\chi^{(2)}
\left(
|E_1|^2
+
|E_2|^2
\right).
$$

positive-frequency成分のcomplex conjugateが、対応するnegative-frequency成分を構成する。


## 3. 2次周波数混合のtensor表現

異方性媒質では、上記のscalar式をtensor contractionとして記述する必要がある。和周波発生の場合、

$$
\omega_3=\omega_1+\omega_2
$$

とおくと、上記のcomplex-amplitude conventionにおけるsum-frequency polarizationは

$$
P_i^{(2)}(\omega_3)
=
2\varepsilon_0
\sum_{j=x,y,z}
\sum_{k=x,y,z}
\chi_{ijk}^{(2)}
(\omega_3;\omega_1,\omega_2)
E_j(\omega_1)
E_k(\omega_2).
$$


### 3.1 1つの出力偏光に対する有用な $3\times3$ matrix

Polarizationの出力成分を1つ固定する。例えば $i=x$ の場合、

$$
\mathbf{\chi}_x^{(2)}
=
\begin{pmatrix}
\chi_{xxx}^{(2)} & \chi_{xxy}^{(2)} & \chi_{xxz}^{(2)}\\
\chi_{xyx}^{(2)} & \chi_{xyy}^{(2)} & \chi_{xyz}^{(2)}\\
\chi_{xzx}^{(2)} & \chi_{xzy}^{(2)} & \chi_{xzz}^{(2)}
\end{pmatrix}.
$$

すると、

$$
P_x^{(2)}(\omega_3)
=
2\varepsilon_0
\mathbf{E}^T(\omega_1)
\mathbf{\chi}_x^{(2)}
\mathbf{E}(\omega_2),
$$

ここで

$$
\mathbf{E}(\omega_m)
=
\begin{pmatrix}
E_x(\omega_m)\\
E_y(\omega_m)\\
E_z(\omega_m)
\end{pmatrix}.
$$

この表現は数値計算において特に簡潔である。

和周波の非線形分極には $E_jE_k$ の積が含まれるため、上式ではHermitian transpose $\dagger$ ではなく通常のtranspose $T$ を用いる。


## 3.2 異なる周波数のSFGにおけるintrinsic permutation symmetry

次の2つの項を考える。

$$
A = E_x(\omega_2)E_y(\omega_1)\chi_{xxy}^{(2)}
$$

および

$$
B = E_y(\omega_2)E_x(\omega_1)\chi_{xyx}^{(2)}.
$$

2つの周波数が異なる場合、すなわち $\omega_1 \neq \omega_2$ のとき、一般には $A \neq B$ である。

その理由は、nonlinear susceptibilityが単にCartesian indexだけでなく、それぞれのindexに対応する周波数にも依存するためである。

第1項 $A$ の周波数argumentを明示すると、

$$
A = \chi_{xxy}^{(2)}(\omega_3;\omega_2,\omega_1)E_x(\omega_2)E_y(\omega_1)
$$

と書ける。

ここで $\omega_3 = \omega_1+\omega_2$ である。

Second-order susceptibilityのintrinsic permutation symmetryは

$$
\chi_{ijk}^{(2)}(\omega_3;\omega_a,\omega_b)
=
\chi_{ikj}^{(2)}(\omega_3;\omega_b,\omega_a)
$$

で与えられる。

したがって現在の場合、

$$
\chi_{xxy}^{(2)}(\omega_3;\omega_2,\omega_1)
=
\chi_{xyx}^{(2)}(\omega_3;\omega_1,\omega_2)
$$

が成立する。

つまりintrinsic permutation symmetryでは、単にCartesian index $x$ と $y$ を交換するのではなく、$(x,\omega_2)$ と $(y,\omega_1)$ という **index-frequency pair全体を同時に交換する**。

したがって $A$ は

$$
A
=
\chi_{xxy}^{(2)}(\omega_3;\omega_2,\omega_1)
E_x(\omega_2)E_y(\omega_1)
$$

$$
=
\chi_{xyx}^{(2)}(\omega_3;\omega_1,\omega_2)
E_y(\omega_1)E_x(\omega_2)
$$

と書くことができる。

したがって、intrinsic permutation symmetryによって保証される正確な関係は

$$
\chi_{xxy}^{(2)}(\omega_3;\omega_2,\omega_1)
E_x(\omega_2)E_y(\omega_1)
=
\chi_{xyx}^{(2)}(\omega_3;\omega_1,\omega_2)
E_y(\omega_1)E_x(\omega_2)
$$

である。

一方、

$$
B
=
\chi_{xyx}^{(2)}(\omega_3;\omega_2,\omega_1)
E_y(\omega_2)E_x(\omega_1)
$$

は $A$ とは異なる周波数・偏光の割り当てを表す。

すなわち $A$ では

- $\omega_2$ 成分が $x$ 偏光
- $\omega_1$ 成分が $y$ 偏光

であるのに対して、$B$ では

- $\omega_2$ 成分が $y$ 偏光
- $\omega_1$ 成分が $x$ 偏光

である。

したがって、$\omega_1 \neq \omega_2$ の一般的なSFGでは $A \neq B$ となり得る。重要なのは、intrinsic permutation symmetryが意味するのは $(j,\omega_1) \leftrightarrow (k,\omega_2)$ の同時交換であり、$j \leftrightarrow k$ だけを独立に交換することではないという点である。

一方、degenerate SHGでは $\omega_1=\omega_2=\omega$ であるため、周波数を交換しても変化せず、

$$
\chi_{ijk}^{(2)}(2\omega;\omega,\omega)
=
\chi_{ikj}^{(2)}(2\omega;\omega,\omega)
$$

となり、最後の2つのCartesian indexについて対称性を利用できる。この対称性により、SHGでは通常の $3\times6$ contracted notationを自然に使用できる。

Kleinman symmetryを適用できる場合には、さらに多くのindex symmetryが得られるが、強いdispersionが存在する領域では自動的に仮定すべきではない。


## 4. SHGのcontracted notation

対称なindex pair $j,k$ に対して、次のように縮約する。

$$
xx\rightarrow1,\qquad
yy\rightarrow2,\qquad
zz\rightarrow3,
$$

$$
yz(=zy)\rightarrow4,\qquad
zx(=xz)\rightarrow5,\qquad
xy(=yx)\rightarrow6.
$$

通常のnonlinear coefficientは

$$
d_{il}
=
\frac{1}{2}\chi_{ijk}^{(2)}
$$

と定義し、$l$ が $jk$ のpairを表す。

単一周波数 $\omega$ のfieldに対して、

$$
\mathbf{q}(\mathbf{E})
=
\begin{pmatrix}
E_x^2\\
E_y^2\\
E_z^2\\
2E_yE_z\\
2E_zE_x\\
2E_xE_y
\end{pmatrix}.
$$

したがってSH polarizationは

$$
\mathbf{P}^{(2)}(2\omega)
=
2\varepsilon_0
\mathbf{d}\,
\mathbf{q}(\mathbf{E})
$$

と書ける。


### 4.1 同一周波数の2つのcoherent beam間のcross term

non-collinear SHGでは、2つのビームを **beam number** で区別する方が明確である。

positive-frequency spatial amplitudeを

$$
\mathbf{E}(\mathbf{r})
=
\mathbf{E}_1(\mathbf{r})
+
\mathbf{E}_2(\mathbf{r})
$$

とし、共通の時間依存性を $e^{-i\omega t}$ とする。

すると、

$$
\mathbf{q}(\mathbf{E}_1+\mathbf{E}_2)
=
\mathbf{q}(\mathbf{E}_1)
+
\mathbf{q}(\mathbf{E}_2)
+
\mathbf{q}_{12},
$$

ここでcross-term vectorは

$$
\mathbf{q}_{12}
=
2
\begin{pmatrix}
E_{1x}E_{2x}\\
E_{1y}E_{2y}\\
E_{1z}E_{2z}\\
E_{1y}E_{2z}+E_{1z}E_{2y}\\
E_{1z}E_{2x}+E_{1x}E_{2z}\\
E_{1x}E_{2y}+E_{1y}E_{2x}
\end{pmatrix}.
$$

したがって、

$$
\mathbf{P}_{12}^{(2)}(2\omega)
=
2\varepsilon_0
\mathbf{d}\,
\mathbf{q}_{12}.
$$

この定義により、しばしば生じるfactor-of-twoの混乱を避けることができる。

もし $\mathbf{q}_{12}$ を先頭のfactor 2なしで定義する場合には、polarization式のprefactorを代わりに $4\varepsilon_0$ としなければならない。


## 5. $\beta$-BBO：結晶対称性と $d$ matrix

$\beta$-$\mathrm{BaB_2O_4}$ (BBO) は非中心対称（non-centrosymmetric）のtrigonal crystalであり、

$$
\text{space group}=R3c,
\qquad
\text{point group}=3m
$$

である。

よく用いられるcrystal-axis conventionでは、contracted second-order matrixは

$$
\mathbf{d}_{\mathrm{BBO}}
=
\begin{pmatrix}
0 & 0 & 0 & 0 & d_{15} & -d_{22}\\
-d_{22} & d_{22} & 0 & d_{15} & 0 & 0\\
d_{31} & d_{31} & d_{33} & 0 & 0 & 0
\end{pmatrix}.
$$

Kleinman symmetryが適用できる場合、

$$
d_{15}=d_{31}.
$$

BBOについてよく引用される係数の大きさは、おおよそ

$$
|d_{22}|
\simeq
2.2\ \mathrm{pm/V},
\qquad
|d_{15}|
=
|d_{31}|
\simeq
0.08\ \mathrm{pm/V}.
$$

各 $d$ coefficientの符号はcrystallographic-axis conventionに依存して変わる場合がある。

上記のmatrix conventionでは、以下のように符号を一貫して使用する必要がある。

$$
d_{21}=-d_{22}.
$$

### 6.1 2つの $y$ 偏光ビームによるcross-term nonlinear polarizationの導出

上で使用した $(x,y,z)$ 軸が、Eq. (15) の **crystal axisと完全に同一の軸** であると仮定する。

2つのfundamental beamは同一周波数 $\omega$ を持ち、$x$-$z$ 平面内を $\pm\theta$ の角度で対称に伝搬するとする。

各電場を

$$
\mathbf{E}_1(\mathbf{r})
=
\begin{pmatrix}
0\\
E_y e^{ik_\omega(\cos\theta\,z+\sin\theta\,x)}\\
0
\end{pmatrix},
$$

$$
\mathbf{E}_2(\mathbf{r})
=
\begin{pmatrix}
0\\
E_y e^{ik_\omega(\cos\theta\,z-\sin\theta\,x)}\\
0
\end{pmatrix}
$$

とする。

したがって、2つのビームを合わせた全fundamental fieldは

$$
\mathbf{E}
=
\mathbf{E}_1+\mathbf{E}_2.
$$

両ビームとも $y$ 偏光成分のみを持つため、

$$
E_x=0,
\qquad
E_z=0,
$$

かつ

$$
E_y
=
E_{1y}+E_{2y}.
$$

SHGのcontracted notationにおけるquadratic field vectorは

$$
\mathbf{q}(\mathbf{E})
=
\begin{pmatrix}
E_x^2\\
E_y^2\\
E_z^2\\
2E_yE_z\\
2E_zE_x\\
2E_xE_y
\end{pmatrix}.
$$

現在の場合 $E_x=E_z=0$ なので、

$$
\mathbf{q}(\mathbf{E})
=
\begin{pmatrix}
0\\
(E_{1y}+E_{2y})^2\\
0\\
0\\
0\\
0
\end{pmatrix}.
$$

第2成分を展開すると、

$$
(E_{1y}+E_{2y})^2
=
E_{1y}^2
+
E_{2y}^2
+
2E_{1y}E_{2y}.
$$

ここで

$$
E_{1y}^2
$$

と

$$
E_{2y}^2
$$

は、それぞれbeam 1とbeam 2が単独で生成するSHG成分であり、

$$
2E_{1y}E_{2y}
$$

が2つのビーム間の **cross-term SHG** に対応する。

したがって、cross termに対応するcontracted field vectorは

$$
\mathbf{q}_{12}
=
\begin{pmatrix}
0\\
2E_{1y}E_{2y}\\
0\\
0\\
0\\
0
\end{pmatrix}.
$$

ここで、2つのfieldの空間位相を代入する。

$$
E_{1y}
=
E_y
e^{ik_\omega(\cos\theta\,z+\sin\theta\,x)},
$$

$$
E_{2y}
=
E_y
e^{ik_\omega(\cos\theta\,z-\sin\theta\,x)}.
$$

したがって2つのfieldの積は

$$
E_{1y}E_{2y}
=
E_y^2
e^{ik_\omega(\cos\theta\,z+\sin\theta\,x)}
e^{ik_\omega(\cos\theta\,z-\sin\theta\,x)}.
$$

指数部をまとめると、

$$
E_{1y}E_{2y}
=
E_y^2
e^{ik_\omega[
2\cos\theta\,z
+
\sin\theta\,x
-
\sin\theta\,x
]}.
$$

$x$ 方向のphase termは相殺されるため、

$$
E_{1y}E_{2y}
=
E_y^2
e^{i2k_\omega\cos\theta\,z}.
$$

したがってcross-term vectorは

$$
\boxed{
\mathbf{q}_{12}
=
2E_y^2
e^{i2k_\omega\cos\theta\,z}
\begin{pmatrix}
0\\
1\\
0\\
0\\
0\\
0
\end{pmatrix}
}
$$

となる。

---

BBOのcontracted nonlinear coefficient matrixを

$$
\mathbf{d}
=
\begin{pmatrix}
0 & 0 & 0 & 0 & d_{15} & -d_{22}\\
-d_{22} & d_{22} & 0 & d_{15} & 0 & 0\\
d_{31} & d_{31} & d_{33} & 0 & 0 & 0
\end{pmatrix}
$$

とする。

SHG nonlinear polarizationは

$$
\mathbf{P}^{(2)}
=
2\varepsilon_0
\mathbf{d}\,
\mathbf{q}
$$

で与えられる。

したがってcross termは

$$
\mathbf{P}_{12}^{(2)}
=
2\varepsilon_0
\mathbf{d}\,
\mathbf{q}_{12}.
$$

先ほど求めた $\mathbf{q}_{12}$ を代入すると、

$$
\mathbf{P}_{12}^{(2)}
=
2\varepsilon_0
\begin{pmatrix}
0 & 0 & 0 & 0 & d_{15} & -d_{22}\\
-d_{22} & d_{22} & 0 & d_{15} & 0 & 0\\
d_{31} & d_{31} & d_{33} & 0 & 0 & 0
\end{pmatrix}
\left[
2E_y^2
e^{i2k_\omega\cos\theta\,z}
\begin{pmatrix}
0\\
1\\
0\\
0\\
0\\
0
\end{pmatrix}
\right].
$$

scalar factorを外に出すと、

$$
\mathbf{P}_{12}^{(2)}
=
4\varepsilon_0
E_y^2
e^{i2k_\omega\cos\theta\,z}
\begin{pmatrix}
0 & 0 & 0 & 0 & d_{15} & -d_{22}\\
-d_{22} & d_{22} & 0 & d_{15} & 0 & 0\\
d_{31} & d_{31} & d_{33} & 0 & 0 & 0
\end{pmatrix}
\begin{pmatrix}
0\\
1\\
0\\
0\\
0\\
0
\end{pmatrix}.
$$

右側vectorでは第2成分のみが1であるため、$d$ matrixの **第2columnのみが選択される**。

すなわち、

$$
\begin{pmatrix}
0 & 0 & 0 & 0 & d_{15} & -d_{22}\\
-d_{22} & d_{22} & 0 & d_{15} & 0 & 0\\
d_{31} & d_{31} & d_{33} & 0 & 0 & 0
\end{pmatrix}
\begin{pmatrix}
0\\
1\\
0\\
0\\
0\\
0
\end{pmatrix}
=
\begin{pmatrix}
0\\
d_{22}\\
d_{31}
\end{pmatrix}.
$$

したがって最終的に、

$$
\boxed{
\mathbf{P}_{12}^{(2)}
=
4\varepsilon_0
E_y^2
e^{i2k_\omega\cos\theta\,z}
\begin{pmatrix}
0\\
d_{22}\\
d_{31}
\end{pmatrix}
}
$$

を得る。

各Cartesian componentは

$$
P_{12,x}^{(2)}=0,
$$

$$
P_{12,y}^{(2)}
=
4\varepsilon_0 d_{22}
E_y^2
e^{i2k_\omega\cos\theta\,z},
$$

$$
P_{12,z}^{(2)}
=
4\varepsilon_0 d_{31}
E_y^2
e^{i2k_\omega\cos\theta\,z}.
$$

ここでspatial phase factor

$$
e^{i2k_\omega\cos\theta\,z}
$$

は、cross-term nonlinear polarizationのwave vectorが

$$
\mathbf{K}_{\mathrm{NL}}
=
\mathbf{k}_1+\mathbf{k}_2
=
\begin{pmatrix}
0\\
0\\
2k_\omega\cos\theta
\end{pmatrix}
$$

であることを意味する。

つまり、対称なnon-collinear geometryでは、cross-term nonlinear polarizationのspatial phaseは2つのfundamental beamのangle bisectorである $z$ 方向に進む。


### 6.2 実際のType-I BBOのcrystal coordinateを反映した計算

6.1節では、laboratory $(x,y,z)$ 軸をEq. (15)の **crystal axisと完全に同一の軸** であると意図的に仮定した。  
したがって6.1節の計算は、その誤った仮定を維持した比較用の計算としてそのまま残す。

実際のType-I BBOでは、laboratory coordinateとcrystal coordinateを区別する必要がある。

ここではEq. (15)の $d$ matrixに対応する標準的な $3m$ crystal-axis conventionを用いて再計算する。

---

#### 6.2.1 Laboratory coordinate

laboratory coordinate $(x,y,z)$ を以下のように定義する。

- $z$: 2つのfundamental beamのangle bisectorであり、cross-term SHの伝搬方向
- $x$: 2つのbeamが交差するplane内のtransverse direction
- $y$: beam-crossing planeに垂直な方向であり、fundamental beamの偏光方向

したがって、2つのfundamental beamのwave vectorは

$$
\mathbf{k}_1
=
k_\omega
\begin{pmatrix}
\sin\theta\\
0\\
\cos\theta
\end{pmatrix},
\qquad
\mathbf{k}_2
=
k_\omega
\begin{pmatrix}
-\sin\theta\\
0\\
\cos\theta
\end{pmatrix}.
$$

2つのbeamによるcross-term nonlinear polarizationのspatial wave vectorは

$$
\mathbf{K}_{\mathrm{NL}}
=
\mathbf{k}_1+\mathbf{k}_2
=
2k_\omega\cos\theta
\begin{pmatrix}
0\\
0\\
1
\end{pmatrix}.
$$

したがって、cross-term SHのnominal propagation directionはlaboratory $z$ 方向である。

---

#### 6.2.2 BBO crystal coordinate $(X,Y,Z)$

Eq. (15)の $d$ matrixで用いるcrystal coordinateを大文字 $(X,Y,Z)$ で表す。

- $Z$: BBOのcrystallographic $c$-axisであり **optic axis**
- $YZ$ plane: $3m$ point groupに属する1つの **crystal mirror plane**
- $X$: そのmirror planeに垂直なaxis

大きな $d_{22}$ を利用するType-I $oo\rightarrow e$ BBOの標準的なorientationでは、propagation planeをcrystal $YZ$ mirror plane内に置く。  
標準的なspherical-angle conventionでは、最適azimuthを

$$
\phi=-\frac{\pi}{2}
$$

と取ることができる。

現在のgeometryではlaboratory $xz$ planeがoptical principal planeであるため、

$$
\boxed{
YZ_{\mathrm{crys}}
=
xz_{\mathrm{lab}}
}
$$

とする。

したがってcrystal $X$ axisはlaboratory $y$ axisと平行である。  
以下では、right-handed coordinateと $\phi=-\pi/2$ conventionを同時に満たすため、

$$
\boxed{
\hat{\mathbf{X}}
=
-\hat{\mathbf{y}}
}
$$

と定義する。

この符号の選択自体は物理的な違いを生じないが、$d_{22}$ の符号と $\phi$ conventionを一貫して保つために必要である。

generated SH wave vector $\mathbf{k}_{2\omega}$ がlaboratory $z$ 方向を向き、optic axis $Z$ と $z$ のなす角を $\alpha$ とする。

すると、crystal basis vectorをlaboratory coordinateで

$$
\boxed{
\hat{\mathbf{X}}
=
\begin{pmatrix}
0\\
-1\\
0
\end{pmatrix},
\qquad
\hat{\mathbf{Y}}
=
\begin{pmatrix}
\cos\alpha\\
0\\
-\sin\alpha
\end{pmatrix},
\qquad
\hat{\mathbf{Z}}
=
\begin{pmatrix}
\sin\alpha\\
0\\
\cos\alpha
\end{pmatrix}
}
$$

と書ける。

実際、

$$
\hat{\mathbf{X}}\times\hat{\mathbf{Y}}
=
\hat{\mathbf{Z}}
$$

を満たすため、right-handed coordinate systemである。

crystal-frame componentをlaboratory frameへ変換するrotation matrixは、各crystal basis vectorをcolumnとして持つ

$$
\boxed{
R
=
\begin{pmatrix}
0 & \cos\alpha & \sin\alpha\\
-1 & 0 & 0\\
0 & -\sin\alpha & \cos\alpha
\end{pmatrix}
}
$$

である。

したがって、

$$
\boxed{
\mathbf{v}_{\mathrm{lab}}
=
R\mathbf{v}_{\mathrm{crys}},
\qquad
\mathbf{v}_{\mathrm{crys}}
=
R^T\mathbf{v}_{\mathrm{lab}}.
}
$$

---

#### 6.2.3 Wave vectorをcrystal frameへ変換

まずgenerated SHの方向を変換すると、

$$
R^T
\begin{pmatrix}
0\\
0\\
1
\end{pmatrix}
=
\begin{pmatrix}
0\\
-\sin\alpha\\
\cos\alpha
\end{pmatrix}.
$$

すなわちcrystal coordinateでは

$$
\boxed{
\hat{\mathbf{k}}_{2\omega,\mathrm{crys}}
=
\begin{pmatrix}
0\\
-\sin\alpha\\
\cos\alpha
\end{pmatrix}
}
$$

である。

したがって $\mathbf{k}_{2\omega}$ はcrystal $YZ$ mirror plane内にあり、$Z$ axisと $\alpha$ の角をなす。  
また $X$ componentが0で $Y<0$ であるため、このorientationは

$$
\phi=-90^\circ
$$

に対応する。

2つのfundamental beamについても

$$
R^T\mathbf{k}_1
=
k_\omega
\begin{pmatrix}
0\\
-\sin(\alpha-\theta)\\
\cos(\alpha-\theta)
\end{pmatrix},
$$

$$
R^T\mathbf{k}_2
=
k_\omega
\begin{pmatrix}
0\\
-\sin(\alpha+\theta)\\
\cos(\alpha+\theta)
\end{pmatrix}.
$$

したがって $\mathbf{k}_1$、$\mathbf{k}_2$、$\mathbf{k}_{2\omega}$ およびoptic axis $Z$ はすべてcrystal $YZ$ mirror plane内にある。

---

#### 6.2.4 Fundamental polarizationはcrystal $X$ 方向のordinary polarization

laboratory frameにおける2つのfundamental fieldは

$$
\mathbf{E}_{1,\mathrm{lab}}
=
E_y
e^{ik_\omega(\cos\theta\,z+\sin\theta\,x)}
\begin{pmatrix}
0\\
1\\
0
\end{pmatrix},
$$

$$
\mathbf{E}_{2,\mathrm{lab}}
=
E_y
e^{ik_\omega(\cos\theta\,z-\sin\theta\,x)}
\begin{pmatrix}
0\\
1\\
0
\end{pmatrix}.
$$

crystal frameへ変換すると、

$$
\mathbf{E}_{m,\mathrm{crys}}
=
R^T\mathbf{E}_{m,\mathrm{lab}}.
$$

ここで

$$
R^T
\begin{pmatrix}
0\\
1\\
0
\end{pmatrix}
=
\begin{pmatrix}
-1\\
0\\
0
\end{pmatrix}.
$$

したがって、

$$
\boxed{
\mathbf{E}_{1,\mathrm{crys}}
=
-E_y
e^{ik_\omega(\cos\theta\,z+\sin\theta\,x)}
\begin{pmatrix}
1\\
0\\
0
\end{pmatrix}
}
$$

および

$$
\boxed{
\mathbf{E}_{2,\mathrm{crys}}
=
-E_y
e^{ik_\omega(\cos\theta\,z-\sin\theta\,x)}
\begin{pmatrix}
1\\
0\\
0
\end{pmatrix}.
}
$$

すなわち、laboratoryでは $y$ 偏光のbeamであるが、Eq. (15)のcrystal coordinateでは **$X$ 偏光のbeam** である。

$$
\boxed{
E_{X,\mathrm{crys}}=-E_y,
\qquad
E_{Y,\mathrm{crys}}=E_{Z,\mathrm{crys}}=0.
}
$$

符号 $-$ はSHGの $E_X^2$ では消える。

また、ordinary polarizationは $\mathbf{k}$ とoptic axis $Z$ が作るprincipal planeに垂直である。  
現在のprincipal planeはcrystal $YZ$ planeなので、ordinary polarizationは正確にcrystal $X$ 方向である。

これが6.1節の誤ったcoordinate identificationとの最も重要な違いである。

---

#### 6.2.5 Crystal frameにおけるcross-term contracted vector

2つのfieldの $X$ componentの積は

$$
E_{1X}E_{2X}
=
(-E_y)(-E_y)
e^{ik_\omega(\cos\theta\,z+\sin\theta\,x)}
e^{ik_\omega(\cos\theta\,z-\sin\theta\,x)}.
$$

したがって、

$$
E_{1X}E_{2X}
=
E_y^2
e^{i2k_\omega\cos\theta\,z}.
$$

全fieldの2乗におけるcross termは

$$
(E_{1X}+E_{2X})^2
\supset
2E_{1X}E_{2X}
$$

なので、

$$
\boxed{
\mathbf{q}_{12,\mathrm{crys}}
=
2E_y^2
e^{i2k_\omega\cos\theta\,z}
\begin{pmatrix}
1\\
0\\
0\\
0\\
0\\
0
\end{pmatrix}.
}
$$

すなわち、実際のType-I orientationでは $d$ matrixの **第2columnではなく第1column** が選択される。

---

#### 6.2.6 BBO $d$ matrixの適用

Eq. (15)のBBO matrixは

$$
\mathbf{d}_{\mathrm{BBO}}
=
\begin{pmatrix}
0 & 0 & 0 & 0 & d_{15} & -d_{22}\\
-d_{22} & d_{22} & 0 & d_{15} & 0 & 0\\
d_{31} & d_{31} & d_{33} & 0 & 0 & 0
\end{pmatrix}.
$$

したがってcross-term nonlinear polarizationは

$$
\mathbf{P}_{12,\mathrm{crys}}^{(2)}
=
2\varepsilon_0
\mathbf{d}_{\mathrm{BBO}}
\mathbf{q}_{12,\mathrm{crys}}.
$$

すなわち、

$$
\mathbf{P}_{12,\mathrm{crys}}^{(2)}
=
4\varepsilon_0
E_y^2
e^{i2k_\omega\cos\theta\,z}
\begin{pmatrix}
0 & 0 & 0 & 0 & d_{15} & -d_{22}\\
-d_{22} & d_{22} & 0 & d_{15} & 0 & 0\\
d_{31} & d_{31} & d_{33} & 0 & 0 & 0
\end{pmatrix}
\begin{pmatrix}
1\\
0\\
0\\
0\\
0\\
0
\end{pmatrix}.
$$

第1columnのみが残るため、

$$
\boxed{
\mathbf{P}_{12,\mathrm{crys}}^{(2)}
=
4\varepsilon_0
E_y^2
e^{i2k_\omega\cos\theta\,z}
\begin{pmatrix}
0\\
-d_{22}\\
d_{31}
\end{pmatrix}.
}
$$

したがってcrystal coordinateでは

$$
P_X^{(2)}=0,
$$

$$
P_Y^{(2)}
=
-4\varepsilon_0 d_{22}
E_y^2
e^{i2k_\omega\cos\theta\,z},
$$

$$
P_Z^{(2)}
=
4\varepsilon_0 d_{31}
E_y^2
e^{i2k_\omega\cos\theta\,z}.
$$

---

#### 6.2.7 Nonlinear polarizationをlaboratory frameへ回転

laboratory frameでは、

$$
\mathbf{P}_{12,\mathrm{lab}}^{(2)}
=
R
\mathbf{P}_{12,\mathrm{crys}}^{(2)}.
$$

したがって、

$$
\mathbf{P}_{12,\mathrm{lab}}^{(2)}
=
4\varepsilon_0
E_y^2
e^{i2k_\omega\cos\theta\,z}
\begin{pmatrix}
0 & \cos\alpha & \sin\alpha\\
-1 & 0 & 0\\
0 & -\sin\alpha & \cos\alpha
\end{pmatrix}
\begin{pmatrix}
0\\
-d_{22}\\
d_{31}
\end{pmatrix}.
$$

行列積を計算すると、

$$
\boxed{
\mathbf{P}_{12,\mathrm{lab}}^{(2)}
=
4\varepsilon_0
E_y^2
e^{i2k_\omega\cos\theta\,z}
\begin{pmatrix}
-d_{22}\cos\alpha+d_{31}\sin\alpha\\
0\\
d_{22}\sin\alpha+d_{31}\cos\alpha
\end{pmatrix}.
}
$$

すなわち、

$$
\boxed{
P_x^{(2)}
=
4\varepsilon_0
\left(
d_{31}\sin\alpha-d_{22}\cos\alpha
\right)
E_y^2
e^{i2k_\omega\cos\theta\,z}
}
$$

であり、

$$
P_y^{(2)}=0,
$$

$$
P_z^{(2)}
=
4\varepsilon_0
\left(
d_{22}\sin\alpha+d_{31}\cos\alpha
\right)
E_y^2
e^{i2k_\omega\cos\theta\,z}.
$$

---

#### 6.2.8 Extraordinary SHに結合するeffective nonlinear coefficient

generated SHはlaboratory $z$ 方向に伝搬し、optic axisはlaboratory $xz$ plane内にある。

したがってType-I $oo\rightarrow e$ のextraordinary output polarization basisはlaboratory $x$ 方向に取ることができる。

$$
\hat{\mathbf{e}}_{2\omega,e}^{\mathrm{lab}}
=
\begin{pmatrix}
1\\
0\\
0
\end{pmatrix}.
$$

これをcrystal coordinateへ変換すると、

$$
\hat{\mathbf{e}}_{2\omega,e}^{\mathrm{crys}}
=
R^T
\hat{\mathbf{e}}_{2\omega,e}^{\mathrm{lab}}
=
\begin{pmatrix}
0\\
\cos\alpha\\
\sin\alpha
\end{pmatrix}.
$$

したがってcrystal-frame nonlinear polarizationとのprojectionは

$$
d_{\mathrm{eff}}
=
\begin{pmatrix}
0&
\cos\alpha&
\sin\alpha
\end{pmatrix}
\begin{pmatrix}
0\\
-d_{22}\\
d_{31}
\end{pmatrix}.
$$

その結果、

$$
\boxed{
d_{\mathrm{eff}}^{ooe}
=
d_{31}\sin\alpha
-
d_{22}\cos\alpha.
}
$$

を得る。

この式は $3m$ crystalの一般的なType-I式

$$
d_{\mathrm{eff}}^{ooe}
=
d_{31}\sin\alpha
-
d_{22}\cos\alpha\sin3\phi
$$

に

$$
\phi=-90^\circ
$$

を適用した標準BBO conventionと一致する。

ただし、文献によって $X,Y$ axisの符号や $d_{22}$ の符号conventionが異なる場合があるため、式全体の符号よりも **同一のaxis conventionを計算の最後まで一貫して使用すること** が重要である。

さらに精密な計算でspatial walk-off angle $\rho$ まで含めると、標準的な表現はおおよそ

$$
d_{\mathrm{eff}}^{ooe}
=
d_{31}\sin(\alpha+\rho)
-
d_{22}\cos(\alpha+\rho)
$$

となる。

---

#### 6.2.9 $d_{22}$ と $d_{31}$ の数値を使用する際の注意

メーカーやデータシートで

$$
|d_{22}|
\simeq
2.2\ \mathrm{pm/V},
\qquad
|d_{31}|
\simeq
0.08\ \mathrm{pm/V}
$$

のように示される値は、しばしば **magnitude** として解釈すべきである。

標準BBO axis conventionでは、$d_{22}$ と $d_{31}$ の相対符号まで一貫して使用する必要がある。

例えば、

$$
d_{31}=+0.08\ \mathrm{pm/V},
\qquad
d_{22}=-2.2\ \mathrm{pm/V}
$$

という符号conventionを選ぶと、

$$
d_{\mathrm{eff}}
=
0.08\sin\alpha
+
2.2\cos\alpha.
$$

$\alpha=29.2^\circ$ のとき、

$$
\boxed{
d_{\mathrm{eff}}
\simeq
1.96\ \mathrm{pm/V}.
}
$$

逆に、すべてのcrystal-axis sign conventionを反転すると $d_{\mathrm{eff}}$ 全体の符号が変わる場合があるが、SHG conversion efficiencyは

$$
I_{2\omega}\propto |d_{\mathrm{eff}}|^2
$$

に依存するため、観測されるintensityは変化しない。

---

#### 6.2.10 6.1節との違い

6.1節では誤った仮定

$$
y_{\mathrm{lab}}
\equiv
Y_{\mathrm{crys}}
$$

を用いたため、$d$ matrixの第2columnを選択し、

$$
\mathbf{P}^{(2)}
\propto
\begin{pmatrix}
0\\
d_{22}\\
d_{31}
\end{pmatrix}
$$

を得た。

しかし、実際のType-I BBOの標準orientationでは、

$$
\boxed{
y_{\mathrm{lab}}
\parallel
X_{\mathrm{crys}}
}
$$

であり、現在のright-handed $\phi=-90^\circ$ conventionでは具体的に

$$
\boxed{
\hat{\mathbf{X}}
=
-\hat{\mathbf{y}}.
}
$$

したがって実際のcross termは $d$ matrixの第1columnを使用し、

$$
\boxed{
\mathbf{P}_{12,\mathrm{crys}}^{(2)}
\propto
\begin{pmatrix}
0\\
-d_{22}\\
d_{31}
\end{pmatrix}.
}
$$

これをlaboratory frameへ回転し、extraordinary outputへprojectionすると、

$$
\boxed{
d_{\mathrm{eff}}^{ooe}
=
d_{31}\sin\alpha
-
d_{22}\cos\alpha
}
$$

を得る。

つまり6.1節の本質的な誤りは、単にoptic axisの傾きを無視したことだけではなく、**laboratory $y$ polarizationをcrystal $Y$ componentとして誤って解釈したこと** にある。

標準Type-I BBOでは、laboratory $y$ polarizationはcrystal $X$ 方向のordinary polarizationに対応する。
