# Introduction

## 1. 비선형 분극

유전체 매질에서 유도 분극은 전기장의 거듭제곱으로 전개할 수 있다. Cartesian tensor notation으로 쓰면,

$$
P_i(t)
=
\varepsilon_0
\left[
\sum_j \chi_{ij}^{(1)} E_j(t)
+
\sum_{jk} \chi_{ijk}^{(2)} E_j(t)E_k(t)
+
\sum_{jkl} \chi_{ijkl}^{(3)} E_j(t)E_k(t)E_l(t)
+\cdots
\right].
$$

간단히 쓰면,

$$
\mathbf{P}
=
\mathbf{P}^{(1)}
+
\mathbf{P}^{(2)}
+
\mathbf{P}^{(3)}
+\cdots.
$$

tensor rank와 SI 단위는 다음과 같다.

- $\chi^{(1)}$: 2차(rank-2) tensor, SI에서 무차원
- $\chi^{(2)}$: 3차(rank-3) tensor, SI 단위는 $\mathrm{m/V}$
- $\chi^{(3)}$: 4차(rank-4) tensor, SI 단위는 $\mathrm{m^2/V^2}$

$\mathbf{P}^{(2)}$는 2차 비선형 분극(second-order nonlinear polarization),  
$\mathbf{P}^{(3)}$는 3차 비선형 분극(third-order nonlinear polarization)이다.

위의 단순한 대수적 형태는 **instantaneous, local response**를 가정한다. 실제 분산 매질은 일반적으로 과거 시간의 전기장에 의존하는 response function으로 기술되며, 주파수 영역에서는 frequency-dependent susceptibility로 표현된다.

따라서 “lossless이면 dispersion이 없다”는 명제는 일반적으로 옳지 않다. 어떤 파장 영역에서 흡수가 거의 없더라도, 그 영역 밖에 존재하는 공명 때문에 굴절률은 여전히 주파수에 따라 변할 수 있다.

비선형성의 크기를 대략 추정할 때 사용할 수 있는 대표적인 스케일은 atomic electric field이다.

$$
E_{\mathrm{at}}
=
\frac{e}{4\pi\varepsilon_0 a_0^2}
\simeq
5.14\times10^{11}\ \mathrm{V/m},
$$

여기서 $a_0$는 Bohr radius이다.

이에 대응하는 진공 중 intensity는

$$
I_{\mathrm{at}}
=
\frac{1}{2}c\varepsilon_0 E_{\mathrm{at}}^2
\simeq
3.51\times10^{16}\ \mathrm{W/cm^2}.
$$

$E\sim E_{\mathrm{at}}$에서 식의 연속된 항들이 비슷한 크기가 된다고 가정하면,

$$
\chi^{(2)}
\sim
\frac{\chi^{(1)}}{E_{\mathrm{at}}},
\qquad
\chi^{(3)}
\sim
\frac{\chi^{(1)}}{E_{\mathrm{at}}^2}.
$$

$\chi^{(1)}\sim1$로 두면 대략

$$
\chi^{(2)}
\sim
2\times10^{-12}\ \mathrm{m/V},
\qquad
\chi^{(3)}
\sim
4\times10^{-24}\ \mathrm{m^2/V^2}.
$$

이 값들은 order-of-magnitude estimate일 뿐이며, 실제 값은 물질과 광학 주파수에 강하게 의존한다.

## 2. 합주파수 및 차주파수 발생

두 개의 단색파 complex field amplitude를 생각하자. 다음 convention을 사용한다.

$$
\widetilde{E}(t)
=
E_1 e^{-i\omega_1 t}
+
E_2 e^{-i\omega_2 t}
+
c.c.
$$

우선 field를 scalar로 취급하면,

$$
\widetilde{P}^{(2)}(t)
=
\varepsilon_0\chi^{(2)}
\widetilde{E}^{\,2}(t).
$$

따라서

$$
\widetilde{P}^{(2)}(t)
=
\varepsilon_0\chi^{(2)}
\left(
E_1 e^{-i\omega_1 t}
+
E_2 e^{-i\omega_2 t}
+
c.c.
\right)^2.
$$

이를 전개하면,

$$
\widetilde{P}^{(2)}(t)
=
\varepsilon_0\chi^{(2)}
\Big[
E_1^2 e^{-2i\omega_1 t}
+
E_2^2 e^{-2i\omega_2 t}
+
2E_1E_2e^{-i(\omega_1+\omega_2)t}
+
2E_1E_2^*e^{-i(\omega_1-\omega_2)t}
+
c.c.
\Big]
$$

$$
\qquad
+
\varepsilon_0\chi^{(2)}
\left(
E_1E_1^*
+
E_2E_2^*
\right).
$$

이 전개에는 second-harmonic, sum-frequency, difference-frequency, zero-frequency 성분이 모두 포함된다.

positive-frequency 성분은

$$
P^{(2)}(2\omega_1)
=
\varepsilon_0\chi^{(2)}E_1^2,
$$

$$
P^{(2)}(2\omega_2)
=
\varepsilon_0\chi^{(2)}E_2^2,
$$

$$
P^{(2)}(\omega_1+\omega_2)
=
2\varepsilon_0\chi^{(2)}E_1E_2,
$$

그리고 $\omega_1>\omega_2$일 때,

$$
P^{(2)}(\omega_1-\omega_2)
=
2\varepsilon_0\chi^{(2)}E_1E_2^*.
$$

zero-frequency 성분은 optical rectification이다.

$$
P^{(2)}(0)
=
2\varepsilon_0\chi^{(2)}
\left(
|E_1|^2
+
|E_2|^2
\right).
$$

positive-frequency 성분의 complex conjugate가 대응하는 negative-frequency 성분을 구성한다.

## 3. 2차 주파수 혼합의 tensor 표현

비등방성 매질에서는 위의 scalar 식을 tensor contraction으로 써야 한다.

$$
\omega_3=\omega_1+\omega_2
$$

라 두면, 위의 complex-amplitude convention에서 sum-frequency polarization은

$$
P_i^{(2)}(\omega_3)
=
2\varepsilon_0
\sum_{j=x,y,z}
\sum_{k=x,y,z}
\chi_{ijk}^{(2)}
(\omega_3;\omega_1,\omega_2)
E_j(\omega_1)
E_k(\omega_2).
$$

### 3.1 하나의 출력 편광에 대한 유용한 $3\times3$ matrix

출력 성분을 하나 고정하자. 예를 들어 $i=x$라면,

$$
\mathbf{\chi}_x^{(2)}
=
\begin{pmatrix}
\chi_{xxx}^{(2)} & \chi_{xxy}^{(2)} & \chi_{xxz}^{(2)}\\
\chi_{xyx}^{(2)} & \chi_{xyy}^{(2)} & \chi_{xyz}^{(2)}\\
\chi_{xzx}^{(2)} & \chi_{xzy}^{(2)} & \chi_{xzz}^{(2)}
\end{pmatrix}.
$$

그러면

$$
P_x^{(2)}(\omega_3)
=
2\varepsilon_0
\mathbf{E}^T(\omega_1)
\mathbf{\chi}_x^{(2)}
\mathbf{E}(\omega_2),
$$

여기서

$$
\mathbf{E}(\omega_m)
=
\begin{pmatrix}
E_x(\omega_m)\\
E_y(\omega_m)\\
E_z(\omega_m)
\end{pmatrix}.
$$

이 형태는 수치 계산에서 특히 간단하다.

비선형 분극에는 $E_jE_k$의 곱이 들어가므로, 위 식에서는 Hermitian transpose $\dagger$가 아니라 일반 transpose $T$를 사용한다.

### 3.2 Permutation symmetry: 중요한 제한

intrinsic permutation symmetry는

$$
\chi_{ijk}^{(2)}
(\omega_3;\omega_1,\omega_2)
=
\chi_{ikj}^{(2)}
(\omega_3;\omega_2,\omega_1)
$$

이다.

즉, 단순히 $j$와 $k$만 교환하는 것이 아니라, 그 인덱스에 대응되는 주파수 $\omega_1$과 $\omega_2$도 함께 교환해야 한다.

따라서 서로 다른 두 주파수에 대해서는 일반적으로

$$
\chi_{ijk}^{(2)}
(\omega_3;\omega_1,\omega_2)
\neq
\chi_{ikj}^{(2)}
(\omega_3;\omega_1,\omega_2)
$$

일 수 있다.

반면 degenerate SHG에서는

$$
\omega_1=\omega_2=\omega
$$

이므로,

$$
\chi_{ijk}^{(2)}(2\omega;\omega,\omega)
=
\chi_{ikj}^{(2)}(2\omega;\omega,\omega).
$$

즉 마지막 두 인덱스 $j,k$를 서로 바꿀 수 있다.

이 대칭성이 일반적인 $3\times6$ contracted notation을 SHG에서 자연스럽게 사용할 수 있게 한다.

Kleinman symmetry를 적용할 수 있는 경우에는 더 많은 index symmetry를 얻을 수 있지만, 강한 dispersion이 있는 영역에서는 이를 자동으로 가정해서는 안 된다.

## 4. SHG의 contracted notation

대칭적인 index pair $j,k$에 대해 다음과 같이 축약한다.

$$
xx\rightarrow1,\qquad
yy\rightarrow2,\qquad
zz\rightarrow3,
$$

$$
yz(=zy)\rightarrow4,\qquad
zx(=xz)\rightarrow5,\qquad
xy(=yx)\rightarrow6.
$$

통상적인 nonlinear coefficient는

$$
d_{il}
=
\frac{1}{2}\chi_{ijk}^{(2)}
$$

로 정의하며, $l$이 $jk$의 pair를 나타낸다.

단일 주파수 $\omega$의 field에 대해,

$$
\mathbf{q}(\mathbf{E})
=
\begin{pmatrix}
E_x^2\\
E_y^2\\
E_z^2\\
2E_yE_z\\
2E_zE_x\\
2E_xE_y
\end{pmatrix}.
$$

따라서 SH polarization은

$$
\mathbf{P}^{(2)}(2\omega)
=
2\varepsilon_0
\mathbf{d}\,
\mathbf{q}(\mathbf{E})
$$

로 쓸 수 있다.

### 4.1 같은 주파수의 두 coherent beam 사이의 cross term

non-collinear SHG에서는 서로 다른 주파수 $\omega_1,\omega_2$보다 **beam number**로 두 빔을 구분하는 편이 더 명확하다.

positive-frequency spatial amplitude를

$$
\mathbf{E}(\mathbf{r})
=
\mathbf{E}_1(\mathbf{r})
+
\mathbf{E}_2(\mathbf{r})
$$

라 두고, 공통 시간 의존성을 $e^{-i\omega t}$로 둔다.

그러면

$$
\mathbf{q}(\mathbf{E}_1+\mathbf{E}_2)
=
\mathbf{q}(\mathbf{E}_1)
+
\mathbf{q}(\mathbf{E}_2)
+
\mathbf{q}_{12},
$$

여기서 cross-term vector는

$$
\mathbf{q}_{12}
=
2
\begin{pmatrix}
E_{1x}E_{2x}\\
E_{1y}E_{2y}\\
E_{1z}E_{2z}\\
E_{1y}E_{2z}+E_{1z}E_{2y}\\
E_{1z}E_{2x}+E_{1x}E_{2z}\\
E_{1x}E_{2y}+E_{1y}E_{2x}
\end{pmatrix}.
$$

따라서

$$
\mathbf{P}_{12}^{(2)}(2\omega)
=
2\varepsilon_0
\mathbf{d}\,
\mathbf{q}_{12}.
$$

위 정의는 자주 발생하는 factor-of-two 혼동을 피하기 위한 것이다.

만약 $\mathbf{q}_{12}$ 내부를 leading factor 2 없이 정의한다면, polarization 식의 prefactor를 대신 $4\varepsilon_0$로 써야 한다.

## 5. $\beta$-BBO: 결정 대칭과 $d$ matrix

$\beta$-$\mathrm{BaB_2O_4}$ (BBO)는 비중심대칭(non-centrosymmetric) trigonal crystal이며,

$$
\text{space group}=R3c,
\qquad
\text{point group}=3m
$$

이다.

자주 사용되는 crystal-axis convention에서 contracted second-order matrix는

$$
\mathbf{d}_{\mathrm{BBO}}
=
\begin{pmatrix}
0 & 0 & 0 & 0 & d_{15} & -d_{22}\\
-d_{22} & d_{22} & 0 & d_{15} & 0 & 0\\
d_{31} & d_{31} & d_{33} & 0 & 0 & 0
\end{pmatrix}.
$$

Kleinman symmetry가 적용 가능한 경우,

$$
d_{15}=d_{31}.
$$

BBO에 대해 흔히 인용되는 계수의 크기는 대략

$$
|d_{22}|
\simeq
2.2\ \mathrm{pm/V},
\qquad
|d_{15}|
=
|d_{31}|
\simeq
0.08\ \mathrm{pm/V}.
$$

각 $d$ coefficient의 부호는 crystallographic-axis convention에 따라 달라질 수 있다.

위 matrix convention에서는

$$
d_{21}=-d_{22}.
$$

따라서 동일한 축 정의를 유지하면서

$$
d_{21}=+2.2\ \mathrm{pm/V}
$$

와

$$
d_{22}=+2.2\ \mathrm{pm/V}
$$

를 동시에 대입할 수는 없다. 그렇게 하려면 crystal axis 또는 sign convention을 다시 정의해야 한다.

## 6. 대칭적인 non-collinear SHG geometry

같은 주파수 $\omega$를 갖는 두 ordinary-polarized fundamental beam이 laboratory $x$-$z$ plane에서 대칭적으로 교차한다고 하자.

두 wave vector는

$$
\mathbf{k}_1
=
k_\omega
(\sin\theta,\,0,\,\cos\theta),
$$

$$
\mathbf{k}_2
=
k_\omega
(-\sin\theta,\,0,\,\cos\theta).
$$

두 field가 laboratory $y$ 방향으로 선형 편광되어 있다고 두면,

$$
\mathbf{E}_1(\mathbf{r})
=
\begin{pmatrix}
0\\
E_y
e^{ik_\omega(\cos\theta\,z+\sin\theta\,x)}\\
0
\end{pmatrix},
$$

$$
\mathbf{E}_2(\mathbf{r})
=
\begin{pmatrix}
0\\
E_y
e^{ik_\omega(\cos\theta\,z-\sin\theta\,x)}\\
0
\end{pmatrix}.
$$

두 $y$ 성분의 곱은

$$
E_{1y}E_{2y}
=
E_y^2
e^{i(\mathbf{k}_1+\mathbf{k}_2)\cdot\mathbf{r}}
=
E_y^2
e^{i2k_\omega\cos\theta\,z}.
$$

따라서 transverse phase

$$
+k_\omega\sin\theta\,x
$$

와

$$
-k_\omega\sin\theta\,x
$$

는 서로 상쇄된다.

그 결과 cross-term nonlinear polarization의 spatial wave vector는

$$
\mathbf{K}_{\mathrm{NL}}
=
\mathbf{k}_1+\mathbf{k}_2
=
(0,\,0,\,2k_\omega\cos\theta).
$$

즉 대칭적인 geometry에서는 cross-term SH signal이 두 fundamental beam의 angle bisector 방향으로 생성될 것으로 기대된다.

### 6.1 Eq. (15)에 직접 대입하는 경우: 무엇을 의미하는가?

위에서 사용한 $(x,y,z)$ 축이 Eq. (15)의 **crystal axis와 정확히 같은 축**이라고 가정하면,

$$
\mathbf{q}_{12}
=
2E_y^2
e^{i2k_\omega\cos\theta z}
\begin{pmatrix}
0\\
1\\
0\\
0\\
0\\
0
\end{pmatrix}.
$$

따라서

$$
\mathbf{P}_{12}^{(2)}
=
4\varepsilon_0
E_y^2
e^{i2k_\omega\cos\theta z}
\begin{pmatrix}
0\\
d_{22}\\
d_{31}
\end{pmatrix}.
$$

이 결과는 **field component와 $d$ matrix가 동일한 coordinate system으로 표현된 경우에만** 대수적으로 올바르다.

### 6.2 실제 phase-matched BBO 실험에서 필요한 중요한 수정

phase-matched BBO 실험에서는 optic axis가 propagation direction에 대해 기울어져 있다.

따라서 laboratory coordinate와 crystal coordinate는 일반적으로 일치하지 않는다.

이 상태에서 laboratory-frame field component를 회전되지 않은 crystal $d$ matrix에 직접 대입하면, nonlinear polarization 방향이나 $d_{\mathrm{eff}}$를 잘못 계산할 수 있다.

crystal-frame vector를 laboratory frame으로 변환하는 rotation matrix를 $R$이라 하자.

먼저 field를 crystal frame으로 변환한다.

$$
\mathbf{E}'_m
=
R^T\mathbf{E}_m.
$$

crystal frame에서 nonlinear polarization을 계산하면,

$$
P_i'^{(2)}
=
\varepsilon_0
\sum_{jk}
\chi_{ijk}'^{(2)}
E_j'E_k'.
$$

그 다음 다시 laboratory frame으로 회전한다.

$$
\mathbf{P}^{(2)}
=
R\mathbf{P}'^{(2)}.
$$

동일하게 full third-rank tensor 자체를 회전할 수도 있다.

$$
d_{ijk}^{\mathrm{lab}}
=
\sum_{pqr}
R_{ip}R_{jq}R_{kr}
d_{pqr}^{\mathrm{crys}}.
$$

Type-I BBO,

$$
o+o\rightarrow e
$$

에서는 최종 coupling을 $2\omega$의 extraordinary eigenpolarization에 projection하여 구한다.

$$
d_{\mathrm{eff}}
\propto
\widehat{\mathbf{e}}_{2\omega,e}
\cdot
\mathbf{P}^{(2)}.
$$

특정 crystallographic-axis convention에서 자주 쓰이는 BBO 식은

$$
d_{\mathrm{eff}}^{ooe}
=
d_{31}\sin\alpha
-
d_{22}\cos\alpha\sin3\phi,
$$

여기서 $\alpha$는 generated-wave vector와 optic axis 사이의 각도이고, $\phi$는 crystal의 azimuthal orientation이다.

이 식은 적절한 azimuth에서 큰 $d_{22}$ 계수가 Type-I effective nonlinearity를 지배할 수 있음을 보여준다.

## 7. BBO Sellmeier 식과 808 nm $\rightarrow$ 404 nm의 경우

$\lambda$를 $\mu\mathrm{m}$ 단위로 사용할 때, 여기서 사용한 refined BBO Sellmeier 식은 다음과 같다.

### Ordinary ray

$$
n_o^2(\lambda)-1
=
\frac{0.90291\lambda^2}{\lambda^2-0.003926}
+
\frac{0.83155\lambda^2}{\lambda^2-0.018786}
+
\frac{0.76536\lambda^2}{\lambda^2-60.01}.
$$

### Extraordinary ray

$$
n_e^2(\lambda)-1
=
\frac{1.151075\lambda^2}{\lambda^2-0.007142}
+
\frac{0.21803\lambda^2}{\lambda^2-0.02259}
+
\frac{0.656\lambda^2}{\lambda^2-263}.
$$

808 nm와 404 nm에서,

$$
n_o(808\,\mathrm{nm})
=
1.66016,
\qquad
n_e(808\,\mathrm{nm})
=
1.54532,
$$

$$
n_o(404\,\mathrm{nm})
=
1.69198,
\qquad
n_e(404\,\mathrm{nm})
=
1.56729.
$$

extraordinary wave의 wave vector가 optic axis와 $\alpha$의 각을 이룰 때,

$$
\frac{1}{n_{e,\mathrm{eff}}^2(\alpha)}
=
\frac{\cos^2\alpha}
{n_o^2(404\,\mathrm{nm})}
+
\frac{\sin^2\alpha}
{n_e^2(404\,\mathrm{nm})}.
$$

### 7.1 Collinear Type-I phase matching

collinear

$$
o(\omega)+o(\omega)\rightarrow e(2\omega)
$$

SHG에서는

$$
n_{e,\mathrm{eff}}(404\,\mathrm{nm},\alpha)
=
n_o(808\,\mathrm{nm}).
$$

이 조건으로부터

$$
\boxed{
\alpha\simeq28.92^\circ
}
$$

를 얻는다.

이는 800-nm-class Type-I BBO SHG에서 사용되는 약 $29^\circ$의 cut angle과 일치한다.

### 7.2 대칭적인 non-collinear phase matching

앞의 symmetric geometry에서 vector phase matching은

$$
\mathbf{k}_{2\omega}
=
\mathbf{k}_1+\mathbf{k}_2
$$

를 요구한다.

따라서

$$
k_{2\omega}
=
2k_\omega\cos\theta.
$$

결국

$$
\boxed{
n_{e,\mathrm{eff}}(404\,\mathrm{nm},\alpha)
=
n_o(808\,\mathrm{nm})\cos\theta
}
$$

가 된다.

여기서 $\theta$는 **crystal 내부의 half-crossing angle**이다.

따라서 crossing angle이 0이 아니면 필요한 phase-matching orientation이 변하며, 최적 $\alpha$는 collinear 값인 $28.92^\circ$보다 커진다.

예를 들어 air에서 external full crossing angle이 $20^\circ$이고, crystal surface normal이 두 beam을 정확히 이등분한다고 하자.

각 beam의 external incidence angle은 $10^\circ$이다.

ordinary fundamental wave에 대해 Snell's law를 적용하면,

$$
\theta_{\mathrm{int}}
=
\sin^{-1}
\left(
\frac{\sin10^\circ}
{n_o(808\,\mathrm{nm})}
\right)
\simeq
6.00^\circ.
$$

이를 위의 non-collinear phase-matching condition과 extraordinary effective index 식에 대입하면,

$$
\boxed{
\alpha
\simeq
33.42^\circ
}
$$

를 얻는다.

즉 single-beam collinear SHG에 최적화된 BBO orientation이 non-collinear cross-term SHG에도 자동으로 최적이라는 보장은 없다.

## 8. 실험에 대한 주요 결론

1. 하나의 출력 편광 성분만 고려할 때는 $3\times3$ tensor slice $\mathbf{\chi}_i^{(2)}$가 가장 직접적인 표현이다. contracted notation은 편리한 표현일 뿐, 물리적으로 필수적인 것은 아니다.

2. degenerate SHG에서는 susceptibility의 마지막 두 index가 대칭이므로 $3\times6$ contracted matrix를 자연스럽게 사용할 수 있다.

3. 같은 주파수의 두 빔을 다룰 때는 $\mathbf{E}(\omega_1)$, $\mathbf{E}(\omega_2)$보다 $\mathbf{E}_1(\omega)$, $\mathbf{E}_2(\omega)$라고 쓰는 편이 좋다. 이렇게 하면 two-beam SHG와 two-color SFG가 혼동되지 않는다.

4. symmetric non-collinear geometry에서 cross term의 phase는

$$
\exp
\left[
i(\mathbf{k}_1+\mathbf{k}_2)\cdot\mathbf{r}
\right]
$$

이며, 그 wave-vector 방향은 두 beam의 angle bisector이다.

5. BBO $d$ matrix는 반드시 electric-field component와 동일한 coordinate system에서 평가해야 한다. cut crystal에서는 generated polarization을 계산하기 전에 field 또는 tensor를 적절히 회전해야 한다.

6. 808 nm $\rightarrow$ 404 nm의 경우, 여기서 사용한 Sellmeier 식은 collinear Type-I phase-matching angle을 약

$$
28.92^\circ
$$

로 준다. finite non-collinear crossing angle은 이 최적값을 이동시킨다.
