# Susceptibility in NLC

## 1. 비선형 분극

유전체 매질에서 유도 분극은 전기장의 거듭제곱으로 전개할 수 있다. Cartesian tensor notation으로 쓰면,

$$
P_i(t)
=
\varepsilon_0
\left[
\sum_j \chi_{ij}^{(1)} E_j(t)
+
\sum_{jk} \chi_{ijk}^{(2)} E_j(t)E_k(t)
+
\sum_{jkl} \chi_{ijkl}^{(3)} E_j(t)E_k(t)E_l(t)
+\cdots
\right].
$$

간단히 쓰면,

$$
\mathbf{P}
=
\mathbf{P}^{(1)}
+
\mathbf{P}^{(2)}
+
\mathbf{P}^{(3)}
+\cdots.
$$

tensor rank와 SI 단위는 다음과 같다.

- $\chi^{(1)}$: 2차(rank-2) tensor, SI에서 무차원
- $\chi^{(2)}$: 3차(rank-3) tensor, SI 단위는 $\mathrm{m/V}$
- $\chi^{(3)}$: 4차(rank-4) tensor, SI 단위는 $\mathrm{m^2/V^2}$

$\mathbf{P}^{(2)}$는 2차 비선형 분극(second-order nonlinear polarization),  
$\mathbf{P}^{(3)}$는 3차 비선형 분극(third-order nonlinear polarization)이다.

위의 단순한 대수적 형태는 **instantaneous, local response**를 가정한다. 실제 분산 매질은 일반적으로 과거 시간의 전기장에 의존하는 response function으로 기술되며, 주파수 영역에서는 frequency-dependent susceptibility로 표현된다.

비선형성의 크기를 대략 추정할 때 사용할 수 있는 대표적인 스케일은 atomic electric field이다.

$$
E_{\mathrm{at}}
=
\frac{e}{4\pi\varepsilon_0 a_0^2}
\simeq
5.14\times10^{11}\ \mathrm{V/m},
$$

여기서 $a_0$는 Bohr radius이다.

이에 대응하는 진공 중 intensity는

$$
I_{\mathrm{at}}
=
\frac{1}{2}c\varepsilon_0 E_{\mathrm{at}}^2
\simeq
3.51\times10^{16}\ \mathrm{W/cm^2}.
$$

$E\sim E_{\mathrm{at}}$에서 식의 연속된 항들이 비슷한 크기가 된다고 가정하면,

$$
\chi^{(2)}
\sim
\frac{\chi^{(1)}}{E_{\mathrm{at}}},
\qquad
\chi^{(3)}
\sim
\frac{\chi^{(1)}}{E_{\mathrm{at}}^2}.
$$

$\chi^{(1)}\sim1$로 두면 대략

$$
\chi^{(2)}
\sim
2\times10^{-12}\ \mathrm{m/V},
\qquad
\chi^{(3)}
\sim
4\times10^{-24}\ \mathrm{m^2/V^2}.
$$

이 값들은 order-of-magnitude estimate일 뿐이며, 실제 값은 물질과 광학 주파수에 강하게 의존한다.

## 2. 합주파수 및 차주파수 발생

두 개의 단색파 complex field amplitude를 생각하자. 다음 convention을 사용한다.

$$
\widetilde{E}(t)
=
E_1 e^{-i\omega_1 t}
+
E_2 e^{-i\omega_2 t}
+
c.c.
$$

우선 field를 scalar로 취급하면,

$$
\widetilde{P}^{(2)}(t)
=
\varepsilon_0\chi^{(2)}
\widetilde{E}^{\,2}(t).
$$

따라서

$$
\widetilde{P}^{(2)}(t)
=
\varepsilon_0\chi^{(2)}
\left(
E_1 e^{-i\omega_1 t}
+
E_2 e^{-i\omega_2 t}
+
c.c.
\right)^2.
$$

이를 전개하면,

$$
\widetilde{P}^{(2)}(t)
=
\varepsilon_0\chi^{(2)}
\Big[
E_1^2 e^{-2i\omega_1 t}
+
E_2^2 e^{-2i\omega_2 t}
+
2E_1E_2e^{-i(\omega_1+\omega_2)t}
+
2E_1E_2^*e^{-i(\omega_1-\omega_2)t}
+
c.c.
\Big]
$$

$$
\qquad
+
\varepsilon_0\chi^{(2)}
\left(
E_1E_1^*
+
E_2E_2^*
\right).
$$

이 전개에는 second-harmonic, sum-frequency, difference-frequency, zero-frequency 성분이 모두 포함된다.

$\omega_1>\omega_2$일 때, positive-frequency 성분은

$$
P^{(2)}(2\omega_1)
=
\varepsilon_0\chi^{(2)}E_1^2,
$$

$$
P^{(2)}(2\omega_2)
=
\varepsilon_0\chi^{(2)}E_2^2,
$$

$$
P^{(2)}(\omega_1+\omega_2)
=
2\varepsilon_0\chi^{(2)}E_1E_2,
$$


$$
P^{(2)}(\omega_1-\omega_2)
=
2\varepsilon_0\chi^{(2)}E_1E_2^*.
$$

zero-frequency 성분은 아래에 표시된것과 같이 optical rectification이다.

$$
P^{(2)}(0)
=
2\varepsilon_0\chi^{(2)}
\left(
|E_1|^2
+
|E_2|^2
\right).
$$

positive-frequency 성분의 complex conjugate는 대응하는 negative-frequency 성분을 구성한다.

## 3. 2차 주파수 혼합의 tensor 표현

비등방성 매질에서는 위의 scalar 식을 tensor contraction으로 써야 한다. 합주파수의 경우

$$
\omega_3=\omega_1+\omega_2
$$

라 두면, 위의 complex-amplitude convention에서 sum-frequency polarization은

$$
P_i^{(2)}(\omega_3)
=
2\varepsilon_0
\sum_{j=x,y,z}
\sum_{k=x,y,z}
\chi_{ijk}^{(2)}
(\omega_3;\omega_1,\omega_2)
E_j(\omega_1)
E_k(\omega_2).
$$



### 3.1 하나의 출력 편광에 대한 유용한 $3\times3$ matrix

Polarization의 출력 성분을 하나 고정하자. 예를 들어 $i=x$라면,

$$
\mathbf{\chi}_x^{(2)}
=
\begin{pmatrix}
\chi_{xxx}^{(2)} & \chi_{xxy}^{(2)} & \chi_{xxz}^{(2)}\\
\chi_{xyx}^{(2)} & \chi_{xyy}^{(2)} & \chi_{xyz}^{(2)}\\
\chi_{xzx}^{(2)} & \chi_{xzy}^{(2)} & \chi_{xzz}^{(2)}
\end{pmatrix}.
$$

그러면

$$
P_x^{(2)}(\omega_3)
=
2\varepsilon_0
\mathbf{E}^T(\omega_1)
\mathbf{\chi}_x^{(2)}
\mathbf{E}(\omega_2),
$$

여기서

$$
\mathbf{E}(\omega_m)
=
\begin{pmatrix}
E_x(\omega_m)\\
E_y(\omega_m)\\
E_z(\omega_m)
\end{pmatrix}.
$$

이 형태는 수치 계산에서 특히 간단하다.

합주파수의 비선형 분극에는 $E_jE_k$의 곱이 들어가므로, 위 식에서는 Hermitian transpose $\dagger$가 아니라 일반 transpose $T$를 사용한다.

## 3.2 서로 다른 주파수의 SFG에서 intrinsic permutation symmetry

다음 두 항을 생각하자.

$$
A = E_x(\omega_2)E_y(\omega_1)\chi_{xxy}^{(2)}
$$

그리고

$$
B = E_y(\omega_2)E_x(\omega_1)\chi_{xyx}^{(2)}
$$

주파수가 서로 다른 경우, 즉 $\omega_1 \neq \omega_2$ 라면 일반적으로 $A \neq B$ 이다.
그 이유는 nonlinear susceptibility가 단순히 Cartesian index에만 의존하는 것이 아니라, 각 index에 대응하는 주파수에도 의존하기 때문이다.

첫 번째 항 $A$를 주파수 argument까지 명시하면,

$$
A = \chi_{xxy}^{(2)}(\omega_3;\omega_2,\omega_1)E_x(\omega_2)E_y(\omega_1)
$$

로 쓸 수 있다.

여기서 $\omega_3 = \omega_1+\omega_2$ 이다.

Second-order susceptibility의 intrinsic permutation symmetry는

$$
\chi_{ijk}^{(2)}(\omega_3;\omega_a,\omega_b)
=
\chi_{ikj}^{(2)}(\omega_3;\omega_b,\omega_a)
$$

로 주어진다.

따라서 현재의 경우,

$$
\chi_{xxy}^{(2)}(\omega_3;\omega_2,\omega_1)
=
\chi_{xyx}^{(2)}(\omega_3;\omega_1,\omega_2)
$$

가 성립한다.
즉 intrinsic permutation symmetry에서는 단순히 Cartesian index $x$와 $y$만 교환하는 것이 아니라, $(x,\omega_2)$ 와 $(y,\omega_1)$ 이라는 **index-frequency pair 전체를 동시에 교환한다.**

따라서 $A$는 다음과 같이 쓸 수 있다.

$$
A
=
\chi_{xxy}^{(2)}(\omega_3;\omega_2,\omega_1)
E_x(\omega_2)E_y(\omega_1)
$$

$$
=
\chi_{xyx}^{(2)}(\omega_3;\omega_1,\omega_2)
E_y(\omega_1)E_x(\omega_2)
$$

따라서 intrinsic permutation symmetry가 보장하는 정확한 관계는

$$
\chi_{xxy}^{(2)}(\omega_3;\omega_2,\omega_1)
E_x(\omega_2)E_y(\omega_1)
=
\chi_{xyx}^{(2)}(\omega_3;\omega_1,\omega_2)
E_y(\omega_1)E_x(\omega_2)
$$

이다.

반면

$$
B
=
\chi_{xyx}^{(2)}(\omega_3;\omega_2,\omega_1)
E_y(\omega_2)E_x(\omega_1)
$$

는 $A$와 다른 주파수-편광 배치를 나타낸다.

즉 $A$에서는

- $\omega_2$ 성분이 $x$ 편광
- $\omega_1$ 성분이 $y$ 편광

인 반면, $B$에서는

- $\omega_2$ 성분이 $y$ 편광
- $\omega_1$ 성분이 $x$ 편광

이다.

따라서 $\omega_1 \neq \omega_2$인 일반적인 SFG에서는 $A \neq B$ 일 수 있다. 핵심적으로 intrinsic permutation symmetry가 의미하는 것은 $(j,\omega_1) \leftrightarrow (k,\omega_2)$ 의 동시 교환이지, $j \leftrightarrow k$ 만을 독립적으로 교환하는 것이 아니다.

반면 degenerate SHG에서는 $\omega_1=\omega_2=\omega$ 이므로 주파수를 교환해도 변화가 없고,

$$
\chi_{ijk}^{(2)}(2\omega;\omega,\omega)
=
\chi_{ikj}^{(2)}(2\omega;\omega,\omega)
$$

가 되어 마지막 두 Cartesian index에 대해 대칭성을 사용할 수 있다.
이 대칭성이 일반적인 $3\times6$ contracted notation을 SHG에서 자연스럽게 사용할 수 있게 한다.

Kleinman symmetry를 적용할 수 있는 경우에는 더 많은 index symmetry를 얻을 수 있지만, 강한 dispersion이 있는 영역에서는 이를 자동으로 가정해서는 안 된다.

## 4. SHG의 contracted notation

대칭적인 index pair $j,k$에 대해 다음과 같이 축약한다.

$$
xx\rightarrow1,\qquad
yy\rightarrow2,\qquad
zz\rightarrow3,
$$

$$
yz(=zy)\rightarrow4,\qquad
zx(=xz)\rightarrow5,\qquad
xy(=yx)\rightarrow6.
$$

통상적인 nonlinear coefficient는

$$
d_{il}
=
\frac{1}{2}\chi_{ijk}^{(2)}
$$

로 정의하며, $l$이 $jk$의 pair를 나타낸다.

단일 주파수 $\omega$의 field에 대해,

$$
\mathbf{q}(\mathbf{E})
=
\begin{pmatrix}
E_x^2\\
E_y^2\\
E_z^2\\
2E_yE_z\\
2E_zE_x\\
2E_xE_y
\end{pmatrix}.
$$

따라서 SH polarization은

$$
\mathbf{P}^{(2)}(2\omega)
=
2\varepsilon_0
\mathbf{d}\,
\mathbf{q}(\mathbf{E})
$$

로 쓸 수 있다.

### 4.1 같은 주파수의 두 coherent beam 사이의 cross term

non-collinear SHG에서는 **beam number**로 두 빔을 구분하는 편이 더 명확하다.

positive-frequency spatial amplitude를

$$
\mathbf{E}(\mathbf{r})
=
\mathbf{E}_1(\mathbf{r})
+
\mathbf{E}_2(\mathbf{r})
$$

라 두고, 공통 시간 의존성을 $e^{-i\omega t}$로 둔다.

그러면

$$
\mathbf{q}(\mathbf{E}_1+\mathbf{E}_2)
=
\mathbf{q}(\mathbf{E}_1)
+
\mathbf{q}(\mathbf{E}_2)
+
\mathbf{q}_{12},
$$

여기서 cross-term vector는

$$
\mathbf{q}_{12}
=
2
\begin{pmatrix}
E_{1x}E_{2x}\\
E_{1y}E_{2y}\\
E_{1z}E_{2z}\\
E_{1y}E_{2z}+E_{1z}E_{2y}\\
E_{1z}E_{2x}+E_{1x}E_{2z}\\
E_{1x}E_{2y}+E_{1y}E_{2x}
\end{pmatrix}.
$$

따라서

$$
\mathbf{P}_{12}^{(2)}(2\omega)
=
2\varepsilon_0
\mathbf{d}\,
\mathbf{q}_{12}.
$$

위 정의는 자주 발생하는 factor-of-two 혼동을 피하기 위한 것이다.

만약 $\mathbf{q}_{12}$ 내부를 leading factor 2 없이 정의한다면, polarization 식의 prefactor를 대신 $4\varepsilon_0$로 써야 한다.

## 5. $\beta$-BBO: 결정 대칭과 $d$ matrix

$\beta$-$\mathrm{BaB_2O_4}$ (BBO)는 비중심대칭(non-centrosymmetric) trigonal crystal이며,

$$
\text{space group}=R3c,
\qquad
\text{point group}=3m
$$

이다.

자주 사용되는 crystal-axis convention에서 contracted second-order matrix는

$$
\mathbf{d}_{\mathrm{BBO}}
=
\begin{pmatrix}
0 & 0 & 0 & 0 & d_{15} & -d_{22}\\
-d_{22} & d_{22} & 0 & d_{15} & 0 & 0\\
d_{31} & d_{31} & d_{33} & 0 & 0 & 0
\end{pmatrix}.
$$

Kleinman symmetry가 적용 가능한 경우,

$$
d_{15}=d_{31}.
$$

BBO에 대해 흔히 인용되는 계수의 크기는 대략

$$
|d_{22}|
\simeq
2.2\ \mathrm{pm/V},
\qquad
|d_{15}|
=
|d_{31}|
\simeq
0.08\ \mathrm{pm/V}.
$$

각 $d$ coefficient의 부호는 crystallographic-axis convention에 따라 달라질 수 있다.

위 matrix convention에서는 아래의 식에서처럼 부호를 사용해야 한다.

$$
d_{21}=-d_{22}.
$$

### 6.1 두 $y$-편광 빔의 cross-term nonlinear polarization 유도

위에서 사용한 $(x,y,z)$ 축이 Eq. (15)의 **crystal axis와 정확히 같은 축**이라고 가정한다.

두 fundamental beam은 동일한 주파수 $\omega$를 가지며, $x$-$z$ 평면에서 $\pm\theta$의 각도로 대칭적으로 전파한다고 하자.

각각의 전기장을

$$
\mathbf{E}_1(\mathbf{r})
=
\begin{pmatrix}
0\\
E_y e^{ik_\omega(\cos\theta\,z+\sin\theta\,x)}\\
0
\end{pmatrix},
$$

$$
\mathbf{E}_2(\mathbf{r})
=
\begin{pmatrix}
0\\
E_y e^{ik_\omega(\cos\theta\,z-\sin\theta\,x)}\\
0
\end{pmatrix}
$$

로 둔다.

따라서 두 빔을 합친 전체 fundamental field는

$$
\mathbf{E}
=
\mathbf{E}_1+\mathbf{E}_2.
$$

두 빔 모두 $y$-편광만 가지고 있으므로,

$$
E_x=0,
\qquad
E_z=0,
$$

이고

$$
E_y
=
E_{1y}+E_{2y}.
$$

SHG의 contracted notation에서 quadratic field vector는

$$
\mathbf{q}(\mathbf{E})
=
\begin{pmatrix}
E_x^2\\
E_y^2\\
E_z^2\\
2E_yE_z\\
2E_zE_x\\
2E_xE_y
\end{pmatrix}.
$$

현재의 경우 $E_x=E_z=0$이므로,

$$
\mathbf{q}(\mathbf{E})
=
\begin{pmatrix}
0\\
(E_{1y}+E_{2y})^2\\
0\\
0\\
0\\
0
\end{pmatrix}.
$$

두 번째 성분을 전개하면

$$
(E_{1y}+E_{2y})^2
=
E_{1y}^2
+
E_{2y}^2
+
2E_{1y}E_{2y}.
$$

여기서

$$
E_{1y}^2
$$

와

$$
E_{2y}^2
$$

는 각각 beam 1과 beam 2가 단독으로 만들어내는 SHG 성분이고,

$$
2E_{1y}E_{2y}
$$

가 두 빔 사이의 **cross-term SHG**에 해당한다.

따라서 cross term에 해당하는 contracted field vector는

$$
\mathbf{q}_{12}
=
\begin{pmatrix}
0\\
2E_{1y}E_{2y}\\
0\\
0\\
0\\
0
\end{pmatrix}.
$$

이제 두 field의 공간 위상을 대입한다.

$$
E_{1y}
=
E_y
e^{ik_\omega(\cos\theta\,z+\sin\theta\,x)},
$$

$$
E_{2y}
=
E_y
e^{ik_\omega(\cos\theta\,z-\sin\theta\,x)}.
$$

따라서 두 field의 곱은

$$
E_{1y}E_{2y}
=
E_y^2
e^{ik_\omega(\cos\theta\,z+\sin\theta\,x)}
e^{ik_\omega(\cos\theta\,z-\sin\theta\,x)}.
$$

지수항을 합치면

$$
E_{1y}E_{2y}
=
E_y^2
e^{ik_\omega[
2\cos\theta\,z
+
\sin\theta\,x
-
\sin\theta\,x
]}.
$$

$x$ 방향의 phase term은 서로 상쇄되므로,

$$
E_{1y}E_{2y}
=
E_y^2
e^{i2k_\omega\cos\theta\,z}.
$$

따라서 cross-term vector는

$$
\boxed{
\mathbf{q}_{12}
=
2E_y^2
e^{i2k_\omega\cos\theta\,z}
\begin{pmatrix}
0\\
1\\
0\\
0\\
0\\
0
\end{pmatrix}
}
$$

가 된다.

---

BBO의 contracted nonlinear coefficient matrix를

$$
\mathbf{d}
=
\begin{pmatrix}
0 & 0 & 0 & 0 & d_{15} & -d_{22}\\
-d_{22} & d_{22} & 0 & d_{15} & 0 & 0\\
d_{31} & d_{31} & d_{33} & 0 & 0 & 0
\end{pmatrix}
$$

라고 하자.

SHG nonlinear polarization은

$$
\mathbf{P}^{(2)}
=
2\varepsilon_0
\mathbf{d}\,
\mathbf{q}
$$

로 주어진다.

따라서 cross term은

$$
\mathbf{P}_{12}^{(2)}
=
2\varepsilon_0
\mathbf{d}\,
\mathbf{q}_{12}.
$$

앞에서 얻은 $\mathbf{q}_{12}$를 대입하면,

$$
\mathbf{P}_{12}^{(2)}
=
2\varepsilon_0
\begin{pmatrix}
0 & 0 & 0 & 0 & d_{15} & -d_{22}\\
-d_{22} & d_{22} & 0 & d_{15} & 0 & 0\\
d_{31} & d_{31} & d_{33} & 0 & 0 & 0
\end{pmatrix}
\left[
2E_y^2
e^{i2k_\omega\cos\theta\,z}
\begin{pmatrix}
0\\
1\\
0\\
0\\
0\\
0
\end{pmatrix}
\right].
$$

scalar term을 밖으로 빼면,

$$
\mathbf{P}_{12}^{(2)}
=
4\varepsilon_0
E_y^2
e^{i2k_\omega\cos\theta\,z}
\begin{pmatrix}
0 & 0 & 0 & 0 & d_{15} & -d_{22}\\
-d_{22} & d_{22} & 0 & d_{15} & 0 & 0\\
d_{31} & d_{31} & d_{33} & 0 & 0 & 0
\end{pmatrix}
\begin{pmatrix}
0\\
1\\
0\\
0\\
0\\
0
\end{pmatrix}.
$$

오른쪽 vector에서 두 번째 성분만 1이므로, $d$ matrix의 **두 번째 column만 선택된다.**

즉,

$$
\begin{pmatrix}
0 & 0 & 0 & 0 & d_{15} & -d_{22}\\
-d_{22} & d_{22} & 0 & d_{15} & 0 & 0\\
d_{31} & d_{31} & d_{33} & 0 & 0 & 0
\end{pmatrix}
\begin{pmatrix}
0\\
1\\
0\\
0\\
0\\
0
\end{pmatrix}
=
\begin{pmatrix}
0\\
d_{22}\\
d_{31}
\end{pmatrix}.
$$

따라서 최종적으로

$$
\boxed{
\mathbf{P}_{12}^{(2)}
=
4\varepsilon_0
E_y^2
e^{i2k_\omega\cos\theta\,z}
\begin{pmatrix}
0\\
d_{22}\\
d_{31}
\end{pmatrix}
}
$$

을 얻는다.

즉 각 Cartesian component는

$$
P_{12,x}^{(2)}=0,
$$

$$
P_{12,y}^{(2)}
=
4\varepsilon_0 d_{22}
E_y^2
e^{i2k_\omega\cos\theta\,z},
$$

$$
P_{12,z}^{(2)}
=
4\varepsilon_0 d_{31}
E_y^2
e^{i2k_\omega\cos\theta\,z}.
$$

여기서 spatial phase factor

$$
e^{i2k_\omega\cos\theta\,z}
$$

는 cross-term nonlinear polarization의 wave vector가

$$
\mathbf{K}_{\mathrm{NL}}
=
\mathbf{k}_1+\mathbf{k}_2
=
\begin{pmatrix}
0\\
0\\
2k_\omega\cos\theta
\end{pmatrix}
$$

임을 의미한다.

즉 대칭적인 non-collinear geometry에서는 cross-term nonlinear polarization의 spatial phase가 두 fundamental beam의 angle bisector인 $z$ 방향으로 진행한다.

### 6.2 실제 Type-I BBO의 crystal coordinate를 반영한 계산

6.1절에서는 의도적으로 laboratory $(x,y,z)$ 축을 Eq. (15)의 **crystal axis와 정확히 같은 축**이라고 가정하였다.  
따라서 6.1절의 계산은 그 잘못된 가정을 유지한 비교용 계산으로 그대로 둔다.

실제 Type-I BBO에서는 laboratory coordinate와 crystal coordinate를 분리해야 한다.

여기서는 Eq. (15)의 $d$ matrix에 대응하는 표준 $3m$ crystal-axis convention을 사용하여 다시 계산한다.

---

#### 6.2.1 Laboratory coordinate

laboratory coordinate $(x,y,z)$를 다음과 같이 정의한다.

- $z$: 두 fundamental beam의 angle bisector이며 cross-term SH의 진행 방향
- $x$: 두 beam이 교차하는 plane 안의 transverse direction
- $y$: beam-crossing plane에 수직인 방향이며 fundamental beam의 편광 방향

따라서 두 fundamental beam의 wave vector는

$$
\mathbf{k}_1
=
k_\omega
\begin{pmatrix}
\sin\theta\\
0\\
\cos\theta
\end{pmatrix},
\qquad
\mathbf{k}_2
=
k_\omega
\begin{pmatrix}
-\sin\theta\\
0\\
\cos\theta
\end{pmatrix}.
$$

두 beam의 cross-term nonlinear polarization이 가지는 spatial wave vector는

$$
\mathbf{K}_{\mathrm{NL}}
=
\mathbf{k}_1+\mathbf{k}_2
=
2k_\omega\cos\theta
\begin{pmatrix}
0\\
0\\
1
\end{pmatrix}.
$$

따라서 cross-term SH의 nominal propagation direction은 laboratory $z$ 방향이다.

---

#### 6.2.2 BBO crystal coordinate $(X,Y,Z)$

Eq. (15)의 $d$ matrix에서 사용하는 crystal coordinate를 대문자 $(X,Y,Z)$로 표시한다.

- $Z$: BBO의 crystallographic $c$-axis이며 **optic axis**
- $YZ$ plane: $3m$ point group의 한 **crystal mirror plane**
- $X$: 그 mirror plane에 수직인 axis

Type-I $oo\rightarrow e$ BBO에서 큰 $d_{22}$를 이용하는 표준적인 orientation에서는 propagation plane을 crystal $YZ$ mirror plane에 둔다.  
표준 spherical-angle convention으로는 최적 azimuth를

$$
\phi=-\frac{\pi}{2}
$$

로 둘 수 있다.

현재 geometry에서는 laboratory $xz$ plane이 바로 optical principal plane이므로,

$$
\boxed{
YZ_{\mathrm{crys}}
=
xz_{\mathrm{lab}}
}
$$

로 둔다.

따라서 crystal $X$ axis는 laboratory $y$ axis와 평행하다.  
아래에서는 right-handed coordinate와 $\phi=-\pi/2$ convention을 동시에 만족시키기 위해

$$
\boxed{
\hat{\mathbf{X}}
=
-\hat{\mathbf{y}}
}
$$

로 정의한다.

이 부호 선택 자체는 물리적인 차이를 만들지 않지만, $d_{22}$의 부호와 $\phi$ convention을 일관되게 유지하는 데 필요하다.

generated SH wave vector $\mathbf{k}_{2\omega}$가 laboratory $z$ 방향이고, optic axis $Z$와 $z$ 사이의 각을 $\alpha$라고 하자.

그러면 crystal basis vectors를 laboratory coordinate에서

$$
\boxed{
\hat{\mathbf{X}}
=
\begin{pmatrix}
0\\
-1\\
0
\end{pmatrix},
\qquad
\hat{\mathbf{Y}}
=
\begin{pmatrix}
\cos\alpha\\
0\\
-\sin\alpha
\end{pmatrix},
\qquad
\hat{\mathbf{Z}}
=
\begin{pmatrix}
\sin\alpha\\
0\\
\cos\alpha
\end{pmatrix}
}
$$

로 둘 수 있다.

실제로

$$
\hat{\mathbf{X}}\times\hat{\mathbf{Y}}
=
\hat{\mathbf{Z}}
$$

이므로 right-handed coordinate이다.

crystal-frame component를 laboratory frame으로 변환하는 rotation matrix는 각 crystal basis vector를 column으로 갖는

$$
\boxed{
R
=
\begin{pmatrix}
0 & \cos\alpha & \sin\alpha\\
-1 & 0 & 0\\
0 & -\sin\alpha & \cos\alpha
\end{pmatrix}
}
$$

이다.

따라서

$$
\boxed{
\mathbf{v}_{\mathrm{lab}}
=
R\mathbf{v}_{\mathrm{crys}},
\qquad
\mathbf{v}_{\mathrm{crys}}
=
R^T\mathbf{v}_{\mathrm{lab}}.
}
$$

---

#### 6.2.3 Wave vector를 crystal frame으로 변환

먼저 generated SH의 방향을 변환하면,

$$
R^T
\begin{pmatrix}
0\\
0\\
1
\end{pmatrix}
=
\begin{pmatrix}
0\\
-\sin\alpha\\
\cos\alpha
\end{pmatrix}.
$$

즉 crystal coordinate에서

$$
\boxed{
\hat{\mathbf{k}}_{2\omega,\mathrm{crys}}
=
\begin{pmatrix}
0\\
-\sin\alpha\\
\cos\alpha
\end{pmatrix}
}
$$

이다.

따라서 $\mathbf{k}_{2\omega}$는 crystal $YZ$ mirror plane 안에 있고, $Z$ axis와 $\alpha$의 각을 이룬다.  
또한 $X$ component가 0이고 $Y<0$이므로 이 orientation은

$$
\phi=-90^\circ
$$

에 해당한다.

두 fundamental beam에 대해서도

$$
R^T\mathbf{k}_1
=
k_\omega
\begin{pmatrix}
0\\
-\sin(\alpha-\theta)\\
\cos(\alpha-\theta)
\end{pmatrix},
$$

$$
R^T\mathbf{k}_2
=
k_\omega
\begin{pmatrix}
0\\
-\sin(\alpha+\theta)\\
\cos(\alpha+\theta)
\end{pmatrix}.
$$

따라서 $\mathbf{k}_1$, $\mathbf{k}_2$, $\mathbf{k}_{2\omega}$ 및 optic axis $Z$는 모두 crystal $YZ$ mirror plane 안에 있다.

---

#### 6.2.4 Fundamental polarization은 crystal $X$ 방향의 ordinary polarization

laboratory frame에서 두 fundamental field는

$$
\mathbf{E}_{1,\mathrm{lab}}
=
E_y
e^{ik_\omega(\cos\theta\,z+\sin\theta\,x)}
\begin{pmatrix}
0\\
1\\
0
\end{pmatrix},
$$

$$
\mathbf{E}_{2,\mathrm{lab}}
=
E_y
e^{ik_\omega(\cos\theta\,z-\sin\theta\,x)}
\begin{pmatrix}
0\\
1\\
0
\end{pmatrix}.
$$

crystal frame으로 변환하면

$$
\mathbf{E}_{m,\mathrm{crys}}
=
R^T\mathbf{E}_{m,\mathrm{lab}}.
$$

여기서

$$
R^T
\begin{pmatrix}
0\\
1\\
0
\end{pmatrix}
=
\begin{pmatrix}
-1\\
0\\
0
\end{pmatrix}.
$$

따라서

$$
\boxed{
\mathbf{E}_{1,\mathrm{crys}}
=
-E_y
e^{ik_\omega(\cos\theta\,z+\sin\theta\,x)}
\begin{pmatrix}
1\\
0\\
0
\end{pmatrix}
}
$$

및

$$
\boxed{
\mathbf{E}_{2,\mathrm{crys}}
=
-E_y
e^{ik_\omega(\cos\theta\,z-\sin\theta\,x)}
\begin{pmatrix}
1\\
0\\
0
\end{pmatrix}.
}
$$

즉 laboratory에서는 $y$-polarized beam이지만, Eq. (15)의 crystal coordinate에서는 **$X$-polarized beam**이다.

$$
\boxed{
E_{X,\mathrm{crys}}=-E_y,
\qquad
E_{Y,\mathrm{crys}}=E_{Z,\mathrm{crys}}=0.
}
$$

부호 $-$는 SHG의 $E_X^2$에서는 사라진다.

또한 ordinary polarization은 $\mathbf{k}$와 optic axis $Z$가 만드는 principal plane에 수직이다.  
현재 principal plane이 crystal $YZ$ plane이므로 ordinary polarization은 정확히 crystal $X$ 방향이다.

이것이 6.1절의 잘못된 coordinate identification과 가장 중요한 차이이다.

---

#### 6.2.5 Crystal frame에서 cross-term contracted vector

두 field의 $X$ component 곱은

$$
E_{1X}E_{2X}
=
(-E_y)(-E_y)
e^{ik_\omega(\cos\theta\,z+\sin\theta\,x)}
e^{ik_\omega(\cos\theta\,z-\sin\theta\,x)}.
$$

따라서

$$
E_{1X}E_{2X}
=
E_y^2
e^{i2k_\omega\cos\theta\,z}.
$$

전체 field의 제곱에서 cross term은

$$
(E_{1X}+E_{2X})^2
\supset
2E_{1X}E_{2X}
$$

이므로,

$$
\boxed{
\mathbf{q}_{12,\mathrm{crys}}
=
2E_y^2
e^{i2k_\omega\cos\theta\,z}
\begin{pmatrix}
1\\
0\\
0\\
0\\
0\\
0
\end{pmatrix}.
}
$$

즉 실제 Type-I orientation에서는 $d$ matrix의 **두 번째 column이 아니라 첫 번째 column**이 선택된다.

---

#### 6.2.6 BBO $d$ matrix 적용

Eq. (15)의 BBO matrix는

$$
\mathbf{d}_{\mathrm{BBO}}
=
\begin{pmatrix}
0 & 0 & 0 & 0 & d_{15} & -d_{22}\\
-d_{22} & d_{22} & 0 & d_{15} & 0 & 0\\
d_{31} & d_{31} & d_{33} & 0 & 0 & 0
\end{pmatrix}.
$$

따라서 cross-term nonlinear polarization은

$$
\mathbf{P}_{12,\mathrm{crys}}^{(2)}
=
2\varepsilon_0
\mathbf{d}_{\mathrm{BBO}}
\mathbf{q}_{12,\mathrm{crys}}.
$$

즉,

$$
\mathbf{P}_{12,\mathrm{crys}}^{(2)}
=
4\varepsilon_0
E_y^2
e^{i2k_\omega\cos\theta\,z}
\begin{pmatrix}
0 & 0 & 0 & 0 & d_{15} & -d_{22}\\
-d_{22} & d_{22} & 0 & d_{15} & 0 & 0\\
d_{31} & d_{31} & d_{33} & 0 & 0 & 0
\end{pmatrix}
\begin{pmatrix}
1\\
0\\
0\\
0\\
0\\
0
\end{pmatrix}.
$$

첫 번째 column만 남으므로,

$$
\boxed{
\mathbf{P}_{12,\mathrm{crys}}^{(2)}
=
4\varepsilon_0
E_y^2
e^{i2k_\omega\cos\theta\,z}
\begin{pmatrix}
0\\
-d_{22}\\
d_{31}
\end{pmatrix}.
}
$$

따라서 crystal coordinate에서

$$
P_X^{(2)}=0,
$$

$$
P_Y^{(2)}
=
-4\varepsilon_0 d_{22}
E_y^2
e^{i2k_\omega\cos\theta\,z},
$$

$$
P_Z^{(2)}
=
4\varepsilon_0 d_{31}
E_y^2
e^{i2k_\omega\cos\theta\,z}.
$$

---

#### 6.2.7 Nonlinear polarization을 laboratory frame으로 회전

laboratory frame에서는

$$
\mathbf{P}_{12,\mathrm{lab}}^{(2)}
=
R
\mathbf{P}_{12,\mathrm{crys}}^{(2)}.
$$

따라서

$$
\mathbf{P}_{12,\mathrm{lab}}^{(2)}
=
4\varepsilon_0
E_y^2
e^{i2k_\omega\cos\theta\,z}
\begin{pmatrix}
0 & \cos\alpha & \sin\alpha\\
-1 & 0 & 0\\
0 & -\sin\alpha & \cos\alpha
\end{pmatrix}
\begin{pmatrix}
0\\
-d_{22}\\
d_{31}
\end{pmatrix}.
$$

행렬곱을 수행하면,

$$
\boxed{
\mathbf{P}_{12,\mathrm{lab}}^{(2)}
=
4\varepsilon_0
E_y^2
e^{i2k_\omega\cos\theta\,z}
\begin{pmatrix}
-d_{22}\cos\alpha+d_{31}\sin\alpha\\
0\\
d_{22}\sin\alpha+d_{31}\cos\alpha
\end{pmatrix}.
}
$$

즉

$$
\boxed{
P_x^{(2)}
=
4\varepsilon_0
\left(
d_{31}\sin\alpha-d_{22}\cos\alpha
\right)
E_y^2
e^{i2k_\omega\cos\theta\,z}
}
$$

이고,

$$
P_y^{(2)}=0,
$$

$$
P_z^{(2)}
=
4\varepsilon_0
\left(
d_{22}\sin\alpha+d_{31}\cos\alpha
\right)
E_y^2
e^{i2k_\omega\cos\theta\,z}.
$$

---

#### 6.2.8 Extraordinary SH에 결합하는 effective nonlinear coefficient

generated SH는 laboratory $z$ 방향으로 진행하고, optic axis는 laboratory $xz$ plane 안에 있다.

따라서 Type-I $oo\rightarrow e$의 extraordinary output polarization basis는 laboratory $x$ 방향으로 둘 수 있다.

$$
\hat{\mathbf{e}}_{2\omega,e}^{\mathrm{lab}}
=
\begin{pmatrix}
1\\
0\\
0
\end{pmatrix}.
$$

이를 crystal coordinate로 변환하면

$$
\hat{\mathbf{e}}_{2\omega,e}^{\mathrm{crys}}
=
R^T
\hat{\mathbf{e}}_{2\omega,e}^{\mathrm{lab}}
=
\begin{pmatrix}
0\\
\cos\alpha\\
\sin\alpha
\end{pmatrix}.
$$

따라서 crystal-frame nonlinear polarization과의 projection은

$$
d_{\mathrm{eff}}
=
\begin{pmatrix}
0&
\cos\alpha&
\sin\alpha
\end{pmatrix}
\begin{pmatrix}
0\\
-d_{22}\\
d_{31}
\end{pmatrix}.
$$

결과적으로

$$
\boxed{
d_{\mathrm{eff}}^{ooe}
=
d_{31}\sin\alpha
-
d_{22}\cos\alpha.
}
$$

이 식은 $3m$ crystal의 일반적인 Type-I 식

$$
d_{\mathrm{eff}}^{ooe}
=
d_{31}\sin\alpha
-
d_{22}\cos\alpha\sin3\phi
$$

에

$$
\phi=-90^\circ
$$

를 적용한 표준 BBO convention과 일치한다.  
단, 문헌마다 $X,Y$ axis의 부호 및 $d_{22}$의 부호 convention이 다를 수 있으므로, 식의 전체 부호보다 **동일한 axis convention을 끝까지 유지하는 것**이 중요하다.

또한 더 정밀한 계산에서 spatial walk-off angle $\rho$까지 포함하면 표준 표현은 대략

$$
d_{\mathrm{eff}}^{ooe}
=
d_{31}\sin(\alpha+\rho)
-
d_{22}\cos(\alpha+\rho)
$$

형태가 된다.

---

#### 6.2.9 $d_{22}$와 $d_{31}$의 수치를 사용할 때의 주의

제조사나 데이터시트에서

$$
|d_{22}|
\simeq
2.2\ \mathrm{pm/V},
\qquad
|d_{31}|
\simeq
0.08\ \mathrm{pm/V}
$$

처럼 주어지는 값은 흔히 **magnitude**로 해석해야 한다.

표준 BBO axis convention에서는 $d_{22}$와 $d_{31}$의 상대 부호까지 일관되게 사용해야 한다.

예를 들어

$$
d_{31}=+0.08\ \mathrm{pm/V},
\qquad
d_{22}=-2.2\ \mathrm{pm/V}
$$

라는 부호 convention을 선택하면,

$$
d_{\mathrm{eff}}
=
0.08\sin\alpha
+
2.2\cos\alpha.
$$

$\alpha=29.2^\circ$이면

$$
\boxed{
d_{\mathrm{eff}}
\simeq
1.96\ \mathrm{pm/V}.
}
$$

반대로 모든 crystal-axis sign convention을 뒤집으면 $d_{\mathrm{eff}}$의 전체 부호가 바뀔 수 있지만, SHG conversion efficiency는

$$
I_{2\omega}\propto |d_{\mathrm{eff}}|^2
$$

에 의존하므로 observable intensity는 변하지 않는다.

---

#### 6.2.10 6.1절과의 차이

6.1절에서는 잘못된 가정

$$
y_{\mathrm{lab}}
\equiv
Y_{\mathrm{crys}}
$$

을 사용했기 때문에 $d$ matrix의 두 번째 column을 선택하여

$$
\mathbf{P}^{(2)}
\propto
\begin{pmatrix}
0\\
d_{22}\\
d_{31}
\end{pmatrix}
$$

을 얻었다.

그러나 실제 Type-I BBO의 표준 orientation에서는

$$
\boxed{
y_{\mathrm{lab}}
\parallel
X_{\mathrm{crys}}
}
$$

이고, 현재의 right-handed $\phi=-90^\circ$ convention에서는 구체적으로

$$
\boxed{
\hat{\mathbf{X}}
=
-\hat{\mathbf{y}}.
}
$$

따라서 실제 cross-term은 $d$ matrix의 첫 번째 column을 사용하며,

$$
\boxed{
\mathbf{P}_{12,\mathrm{crys}}^{(2)}
\propto
\begin{pmatrix}
0\\
-d_{22}\\
d_{31}
\end{pmatrix}.
}
$$

이를 laboratory frame으로 회전하고 extraordinary output에 projection하면,

$$
\boxed{
d_{\mathrm{eff}}^{ooe}
=
d_{31}\sin\alpha
-
d_{22}\cos\alpha
}
$$

를 얻는다.

즉 6.1절의 핵심 오류는 단순히 optic axis가 기울어져 있다는 사실을 무시한 것뿐 아니라,  
**laboratory $y$-polarization을 crystal $Y$ component로 잘못 해석했다는 것**이다.

표준 Type-I BBO에서는 laboratory $y$-polarization이 crystal $X$ ordinary polarization에 대응한다.
