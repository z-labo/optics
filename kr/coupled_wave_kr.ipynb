## 비선형 광학 매질의 파동방정식

### 1–2. Maxwell 방정식으로부터 일반 파동방정식 유도

Ampère 법칙을 Faraday 법칙에 대입하는 방식으로 파동방정식을 유도한다.

$$
\vec{\nabla}\times\vec{\nabla}\times\widetilde{\vec E}
=
-\mu_0\partial_t^2\widetilde{\vec D}
$$

$$
\widetilde{\vec D}
=
\varepsilon_0\widetilde{\vec E}
+
\widetilde{\vec P},
\qquad
\mu_0=\frac{1}{\varepsilon_0c^2}
$$

$$
\vec{\nabla}\times\vec{\nabla}\times\widetilde{\vec E}
+
\frac{1}{c^2}\partial_t^2\widetilde{\vec E}
=
-\frac{1}{\varepsilon_0c^2}\partial_t^2\widetilde{\vec P}
\tag{5}
$$


식 (5)는 여기에서 사용하는 고전 비선형 광학의 가장 일반적인 파동방정식 형태이다.

### 3–4. 평면파 조건과 분극의 선형/비선형 분리

광학에서는
$\vec{\nabla}\times\vec{\nabla}\times\widetilde{\vec E}$
를 Laplacian으로 바꾸고 싶다. 그러나 일반적으로 전기장의 발산은 0이 아니다. 평면파에 대해서는

$$
\nabla\cdot\vec E=0
$$

을 사용할 수 있다. 따라서

$$
-\nabla^2\widetilde{\vec E}
+
\frac{1}{c^2}\partial_t^2\widetilde{\vec E}
=
-\frac{1}{\varepsilon_0c^2}\partial_t^2\widetilde{\vec P}
\tag{6}
$$

$$
\widetilde{\vec P}
=
\widetilde{\vec P}^{(1)}
+
\widetilde{\vec P}^{NL}
\tag{7}
$$

$$
-\nabla^2\widetilde{\vec E}
+
\frac{1}{c^2}\partial_t^2
\left(
\widetilde{\vec E}
+
\frac{1}{\varepsilon_0}\widetilde{\vec P}^{(1)}
\right)
=
-\frac{1}{\varepsilon_0c^2}
\partial_t^2\widetilde{\vec P}^{NL}
\tag{8}
$$

여기서
$$
\widetilde{\vec D}^{(1)}
=
\varepsilon_0\widetilde{\vec E}
+
\widetilde{\vec P}^{(1)},
$$

이를 이용하면
$$
-\nabla^2\widetilde{\vec E}
+
\frac{1}{\varepsilon_0c^2}
\partial_t^2\widetilde{\vec D}^{(1)}
=
-\frac{1}{\varepsilon_0c^2}
\partial_t^2\widetilde{\vec P}^{NL}.
\tag{9}
$$

### 5–6. 단순한 경우: 무손실·무분산 매질

무손실이고 분산이 없는 매질에서는 선형 응답이 순간적이라고 가정한다.

$$
\widetilde{\vec D}^{(1)}(t)
=
\varepsilon_0\varepsilon\,
\widetilde{\vec E}(t)
\tag{10}
$$

식 (10)을 식 (9)에 대입하면,
$$
-\nabla^2\widetilde{\vec E}
+
\frac{\varepsilon^{(1)}}{c^2}
\partial_t^2\widetilde{\vec E}
=
-\frac{1}{\varepsilon_0c^2}
\partial_t^2\widetilde{\vec P}^{NL}.
\tag{11}
$$

일반적으로,
$$
D(t)\neq E(t),
\qquad
D(\omega)\sim E(\omega).
$$


식 (11)은 오른쪽 항, 즉 비선형 분극이 소스 항으로 작용하는 파동방정식이다.

### 7–11. 일반적인 경우: 분산 매질과 단색파 성분

분산 매질에서는 시간영역에서 식 (10)이 성립하지 않는다. 그러나 주파수 영역에서는 선형 전기변위와 전기장이 주파수 의존 유전응답으로 연결된다.

2차 비선형 과정은 세 개의 파를 결합하고, 3차 비선형 과정은 네 개의 파를 결합한다. 실제로 결합되는 파의 수는 적기 때문에 전체 장을 단색파 성분들의 합으로 나타낼 수 있다.

$$
\widetilde{\vec D}^{(1)}(\omega)
=
\varepsilon_0\varepsilon^{(1)}(\omega)
\widetilde{\vec E}(\omega)
$$

$$
\widetilde{\vec E}(\vec r,t)
=
\sum_n' \widetilde{\vec E}_n(\vec r,t)
$$

$$
\widetilde{\vec D}(\vec r,t)
=
\sum_n' \widetilde{\vec D}_n(\vec r,t)
$$

$$
\widetilde{\vec P}^{NL}(\vec r,t)
=
\sum_n' \widetilde{\vec P}^{NL}_n(\vec r,t)
\tag{13}
$$

$\varepsilon^{(1)}\in\mathbb{R}$라고 가정하고 주파수 성분을 식 (9)에 대입하면,
$$
-\nabla^2\widetilde{\vec E}_n
+
\frac{\varepsilon^{(1)}(\omega_n)}{c^2}
\partial_t^2\widetilde{\vec E}_n
=
-\frac{1}{\varepsilon_0c^2}
\partial_t^2\widetilde{\vec P}^{NL}_n.
\tag{15}
$$

단색파에 대해서,
$$
\widetilde{\vec E}_n(\vec r,t)
=
\vec E_n(\vec r)e^{-i\omega_nt}
+c.c.,
\qquad
\vec E_n\in\mathbb C^3,
$$

$$
\widetilde{\vec D}_n(\vec r,t)
=
\vec D_n(\vec r)e^{-i\omega_nt}
+c.c.,
\qquad
\vec D_n\in\mathbb C^3,
$$

$$
\widetilde{\vec P}_n(\vec r,t)
=
\vec P_n(\vec r)e^{-i\omega_nt}
+c.c.,
\qquad
\vec P_n\in\mathbb C^3.
\tag{16}
$$

공간 의존성과 시간 의존성을 분리했으므로,
$$
\partial_t^2\sim-\omega^2,
$$

따라서
$$
\nabla^2\widetilde{\vec E}_n(\vec r)
+
\frac{\varepsilon^{(1)}(\omega_n)\omega_n^2}{c^2}
\widetilde{\vec E}_n
=
-\frac{\omega_n^2}{\varepsilon_0c^2}
\widetilde{\vec P}^{NL}_n.
\tag{17}
$$


식 (15)는 식 (11)과 같은 형태를 가지면서, 물질의 응답이 주파수에 의존하는 경우에도 사용할 수 있다.

## 합주파수 생성(SFG)의 결합파 방정식

### 1–4. 장 방정식과 완만하게 변하는 진폭에 대한 Ansatz

이제 앞에서 얻은 파동방정식을 구체적인 비선형 상호작용인 합주파수 생성(SFG)에 적용한다.

SFG에서는,
$$
\omega_3=\omega_1+\omega_2.
$$

식 (15)의 $\omega_3$ 성분은 다음과 같다.
$$
-\nabla^2\widetilde{\vec E}_3
+
\frac{n_3^2}{c^2}
\partial_t^2\widetilde{\vec E}_3
=
-\frac{1}{\varepsilon_0c^2}
\partial_t^2\widetilde{\vec P}^{NL}_3.
\tag{20}
$$

오른쪽 항을 무시할 수 있다면 일반적인 파동방정식의 해는
$$
\widetilde E_3(z,t)
=
A_3e^{i(k_3z-\omega_3t)}
+c.c.,
$$

$$
k_3=\frac{n_3\omega_3}{c}.
\tag{21}
$$

비선형 소스가 작지만 0은 아닐 때 다음 Ansatz를 사용한다.
$$
\widetilde E_3(z,t)
=
A_3(z)e^{i(k_3z-\omega_3t)}
+c.c.
$$


결정 내부를 진행하면서 생성파의 진폭 $A_3(z)$가 증가할 것으로 예상되므로, 진폭이 $z$에 의존하도록 두는 것은 물리적으로 자연스럽다.

### 5–6. 비선형 소스 분극

비선형 소스 항은 다음과 같이 쓴다.
$$
\widetilde P_3(z,t)
=
P_3(z)e^{-i\omega_3t}
+c.c.,
$$

$$
P_3
=
4\varepsilon_0d_{\mathrm{eff}}E_1E_2,
\qquad
2d_{\mathrm{eff}}\sim\chi^{(2)}.
\tag{24}
$$

또한,
$$
E_i(z,t)
=
E_ie^{-i\omega_it}+c.c.,
\qquad
E_i=A_ie^{ik_iz}.
\tag{25}
$$

따라서,
$$
P_3
=
4\varepsilon_0d_{\mathrm{eff}}
A_1A_2e^{i(k_1+k_2)z}.
\tag{26}
$$

### 7. 파동방정식에 대입

장의 Ansatz와 식 (26)을 식 (20)에 대입한다.

첫 번째 항:
$$
-\nabla^2\widetilde{\vec E}_3
=
-\partial_z^2
\left[
A_3(z)e^{i(k_3z-\omega_3t)}+c.c.
\right]
$$

$$
=
-
\left[
\frac{\partial^2A_3}{\partial z^2}
+
2ik_3\frac{\partial A_3}{\partial z}
-
k_3^2A_3
\right]
e^{i(k_3z-\omega_3t)}
+c.c.
$$

두 번째 항:
$$
\frac{n_3^2}{c^2}\partial_t^2\widetilde{\vec E}_3
=
-\frac{n_3^2\omega_3^2}{c^2}
\left[
A_3e^{i(k_3z-\omega_3t)}+c.c.
\right].
$$

세 번째 항:
$$
-\frac{1}{\varepsilon_0c^2}
\partial_t^2\widetilde{\vec P}^{NL}_3
=
\frac{4\varepsilon_0d_{\mathrm{eff}}\omega_3^2}
{\varepsilon_0c^2}
A_1A_2
e^{i[(k_1+k_2)z-\omega_3t]}
+c.c.
$$

### 10. 각 항을 합치기

각 항을 합치면,
$$
\left[
-\frac{\partial^2A_3}{\partial z^2}
-2ik_3\frac{\partial A_3}{\partial z}
+k_3^2A_3
-\frac{n_3^2\omega_3^2}{c^2}A_3
\right]
e^{i(k_3z-\omega_3t)}
+c.c.
=
\frac{4\varepsilon_0d_{\mathrm{eff}}\omega_3^2}
{\varepsilon_0c^2}
A_1A_2
e^{i[(k_1+k_2)z-\omega_3t]}
+c.c.
$$

여기서
$$
k_3^2A_3-\frac{n_3^2\omega_3^2}{c^2}A_3=0,
$$
이고, 또한 $e^{-i\omega t}$와 $e^{i\omega t}$는 선형 독립이므로 $c.c.$ 항은 별도로 취급할 수 있다. $e^{-i\omega_3t}$ 로 나누면
$$
-
\left[
\frac{\partial^2A_3}{\partial z^2}
+
2ik_3\frac{\partial A_3}{\partial z}
\right]
e^{ik_3z}
=
\frac{4\varepsilon_0d_{\mathrm{eff}}\omega_3^2}
{\varepsilon_0c^2}
A_1A_2e^{i(k_1+k_2)z}.
\tag{27}
$$

### 11–12. 완만한 진폭 근사 (SVAA)

완만한 진폭 근사(SVAA)에서는 두 번째 공간 미분항을 무시한다:
$$
\frac{\partial^2A_3}{\partial z^2}
\ll
k_3\frac{\partial A_3}{\partial z}.
$$

그러면
$$
-2ik_3\frac{\partial A_3}{\partial z}e^{ik_3z}
=
\frac{4\varepsilon_0d_{\mathrm{eff}}\omega_3^2}
{\varepsilon_0c^2}
A_1A_2e^{i(k_1+k_2)z},
$$

따라서
$$
\frac{\partial A_3}{\partial z}
=
i\frac{2d_{\mathrm{eff}}\omega_3^2}{k_3c^2}
A_1A_2e^{i(k_1+k_2-k_3)z}.
$$

다음과 같이 정의한다:
$$
\Delta k=k_1+k_2-k_3,
\qquad
k_3=\frac{n_3\omega_3}{c}.
$$

원 강의 노트에 적힌 식을 그대로 따르면,
$$
\frac{\partial A_3}{\partial z}
=
i\frac{2d_{\mathrm{eff}}\omega_3^2}{n_3c^2}
A_1A_2e^{i\Delta kz}.
\tag{28}
$$

### 13. 세 파에 대한 결합방정식

같은 절차를 $A_1$과 $A_2$에도 적용한다. 다음 관계를 사용한다:
$$
\omega_1=\omega_3-\omega_2,
\qquad
\omega_2=\omega_3-\omega_1.
$$

비선형 분극은
$$
P_1
=
4\varepsilon_0d_{\mathrm{eff}}
A_3A_2^*e^{i(k_3-k_2)z},
$$

$$
P_2
=
4\varepsilon_0d_{\mathrm{eff}}
A_3A_1^*e^{i(k_3-k_1)z}.
$$

따라서,
$$
\frac{\partial A_1}{\partial z}
=
i\frac{2d_{\mathrm{eff}}\omega_1^2}{k_1c^2}
A_3(z)A_2^*(z)e^{-i\Delta kz},
$$

$$
\frac{\partial A_2}{\partial z}
=
i\frac{2d_{\mathrm{eff}}\omega_2^2}{k_2c^2}
A_3(z)A_1^*(z)e^{-i\Delta kz},
$$

$$
\frac{\partial A_3}{\partial z}
=
i\frac{2d_{\mathrm{eff}}\omega_3^2}{k_3c^2}
A_1(z)A_2(z)e^{i\Delta kz}.
$$
